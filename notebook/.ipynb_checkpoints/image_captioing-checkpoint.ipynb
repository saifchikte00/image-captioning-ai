```python
from collections import defaultdict

# Correct paths
images_path = "../dataset/Flicker8k_Dataset"
captions_file = "../dataset/Flickr8k_text/Flickr8k.token.txt"

# Dictionary: image_name -> list of captions
mapping = defaultdict(list)

with open(captions_file, 'r', encoding='utf-8') as f:
    for line in f:
        line = line.strip()
        
        if len(line) == 0:
            continue
        
        # Split into image_id and caption
        tokens = line.split('\t')
        
        if len(tokens) != 2:
            continue
        
        image_caption_id, caption = tokens
        
        # Remove #0, #1, #2... from image name
        image_name = image_caption_id.split('#')[0]
        
        # Store caption
        mapping[image_name].append(caption)

print("✅ Total images with captions:", len(mapping))

# Show sample
sample_image = list(mapping.keys())[0]
print("\nSample image:", sample_image)
print("Captions:")
for cap in mapping[sample_image]:
    print("-", cap)
```

```python
import string
import os

# Keep only captions for images that actually exist in the image folder
valid_images = set(os.listdir(images_path))

mapping = {img: caps for img, caps in mapping.items() if img in valid_images}

print("✅ Valid images with captions after filtering:", len(mapping))

def clean_caption(caption):
    # Convert to lowercase
    caption = caption.lower()
    
    # Remove punctuation
    caption = caption.translate(str.maketrans('', '', string.punctuation))
    
    # Split into words
    words = caption.split()
    
    # Keep only alphabetic words with length > 1
    words = [word for word in words if len(word) > 1 and word.isalpha()]
    
    # Add start and end tokens
    caption = 'startseq ' + ' '.join(words) + ' endseq'
    
    return caption

# Clean all captions
for image_name, captions in mapping.items():
    for i in range(len(captions)):
        captions[i] = clean_caption(captions[i])

print("✅ Captions cleaned successfully")

# Show cleaned sample
sample_image = list(mapping.keys())[0]
print("\nSample cleaned captions for:", sample_image)
for cap in mapping[sample_image]:
    print("-", cap)
    
```

```python
import os
import pickle
import numpy as np
from tqdm import tqdm
from PIL import Image

from tensorflow.keras.applications.vgg16 import VGG16, preprocess_input
from tensorflow.keras.preprocessing.image import img_to_array
from tensorflow.keras.models import Model
```

```python
# Load pre-trained VGG16 model
base_model = VGG16()

# Remove final classification layer
feature_extractor = Model(
    inputs=base_model.inputs,
    outputs=base_model.layers[-2].output
)

print("✅ VGG16 feature extractor loaded successfully")
print("Output shape:", feature_extractor.output_shape)
```

```python
def extract_features_from_image(image_path):
    try:
        # Load image
        image = Image.open(image_path)
        image = image.convert("RGB")
        image = image.resize((224, 224))
        
        # Convert image to array
        image = img_to_array(image)
        image = np.expand_dims(image, axis=0)
        
        # Preprocess for VGG16
        image = preprocess_input(image)
        
        # Extract feature
        feature = feature_extractor.predict(image, verbose=0)
        
        return feature[0]   # shape = (4096,)
    
    except Exception as e:
        print(f"❌ Error processing {image_path}: {e}")
        return None
```

```python
# Correct image folder path (your actual folder name)
images_path = "../dataset/Flicker8k_Dataset"

# List all image files
image_files = os.listdir(images_path)

print("Total images to process:", len(image_files))

# Dictionary: image_name -> feature vector
features = {}

for image_name in tqdm(image_files):
    image_path = os.path.join(images_path, image_name)
    
    feature = extract_features_from_image(image_path)
    
    if feature is not None:
        features[image_name] = feature

print("\n✅ Feature extraction completed")
print("Total extracted features:", len(features))
```

```python
import numpy as np
from tensorflow.keras.utils import to_categorical, Sequence
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, LSTM, Embedding, Dropout, add
from tensorflow.keras.callbacks import ModelCheckpoint
```

```python
# Train images file path
train_file = "../dataset/Flickr8k_text/Flickr_8k.trainImages.txt"

# Load train image names
with open(train_file, 'r', encoding='utf-8') as f:
    train_images = f.read().strip().split('\n')

# Keep only images that exist in features and mapping
train_images = [img for img in train_images if img in mapping and img in features]

print("✅ Total training images:", len(train_images))
print("Sample train images:", train_images[:5])
```

```python
class DataGenerator(Sequence):
    def __init__(self, image_ids, mapping, features, tokenizer, max_length, vocab_size, batch_size=32):
        self.image_ids = image_ids
        self.mapping = mapping
        self.features = features
        self.tokenizer = tokenizer
        self.max_length = max_length
        self.vocab_size = vocab_size
        self.batch_size = batch_size
        self.indexes = np.arange(len(self.image_ids))

    def __len__(self):
        return int(np.ceil(len(self.image_ids) / self.batch_size))

    def __getitem__(self, index):
        batch_indexes = self.indexes[index * self.batch_size:(index + 1) * self.batch_size]
        batch_image_ids = [self.image_ids[i] for i in batch_indexes]

        X1, X2, y = [], [], []

        for image_id in batch_image_ids:
            captions = self.mapping[image_id]
            feature = self.features[image_id]

            for caption in captions:
                seq = self.tokenizer.texts_to_sequences([caption])[0]

                for i in range(1, len(seq)):
                    in_seq = seq[:i]
                    out_seq = seq[i]

                    in_seq = pad_sequences([in_seq], maxlen=self.max_length)[0]
                    out_seq = to_categorical([out_seq], num_classes=self.vocab_size)[0]

                    X1.append(feature)
                    X2.append(in_seq)
                    y.append(out_seq)

        return (np.array(X1), np.array(X2)), np.array(y)
```

```python
# Create generator
train_generator = DataGenerator(
    image_ids=train_images,
    mapping=mapping,
    features=features,
    tokenizer=tokenizer,
    max_length=max_length,
    vocab_size=vocab_size,
    batch_size=32
)

print("✅ Train generator created successfully")
print("Steps per epoch:", len(train_generator))
```

```python
# -----------------------------
# Feature extractor branch
# -----------------------------
inputs1 = Input(shape=(4096,))
fe1 = Dropout(0.4)(inputs1)
fe2 = Dense(256, activation='relu')(fe1)

# -----------------------------
# Sequence model branch
# -----------------------------
inputs2 = Input(shape=(max_length,))
se1 = Embedding(vocab_size, 256, mask_zero=True)(inputs2)
se2 = Dropout(0.4)(se1)
se3 = LSTM(256)(se2)

# -----------------------------
# Decoder
# -----------------------------
decoder1 = add([fe2, se3])
decoder2 = Dense(256, activation='relu')(decoder1)
outputs = Dense(vocab_size, activation='softmax')(decoder2)

# -----------------------------
# Final model
# -----------------------------
model = Model(inputs=[inputs1, inputs2], outputs=outputs)

model.compile(loss='categorical_crossentropy', optimizer='adam')

print("✅ Model built successfully")
model.summary()
```

```python
# Save best model automatically
checkpoint = ModelCheckpoint(
    "../model/image_caption_model.keras",
    monitor='loss',
    save_best_only=True,
    verbose=1
)

# Train model
history = model.fit(
    train_generator,
    epochs=10,
    callbacks=[checkpoint],
    verbose=1
)
```

```python
import os
import torch
import matplotlib.pyplot as plt
import PIL.Image as PILImage

from transformers import BlipProcessor, BlipForConditionalGeneration
```

```python
# Device setup
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)

# Load BLIP processor and model
processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-base").to(device)

print("✅ BLIP model loaded successfully!")
```

```python
def automatic_caption_generation_blip(image_path, show_image=True, max_length=35, num_beams=5):
    """
    Automatic Caption Generation using BLIP (best practical version).
    """
    if not os.path.exists(image_path):
        raise FileNotFoundError(f"❌ Image not found: {image_path}")

    # Open image safely
    image = PILImage.open(image_path).convert("RGB")

    # Show image
    if show_image:
        plt.figure(figsize=(8, 6))
        plt.imshow(image)
        plt.axis("off")
        plt.title("Input Image")
        plt.show()

    # Prepare inputs
    print("⏳ Generating caption...")
    inputs = processor(images=image, return_tensors="pt").to(device)

    # Generate caption
    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_length=max_length,
            num_beams=num_beams,
            early_stopping=True
        )

    # Decode
    caption = processor.decode(output[0], skip_special_tokens=True)
    caption = caption.strip().capitalize()

    print("\n✅ Automatic Generated Caption:")
    print(caption)

    
```

```python
from tensorflow.keras.models import Model

# Remove final classification layer and use fc2 output
model = Model(inputs=base_model.inputs, outputs=base_model.layers[-2].output)

print("Feature extractor created successfully!")
print("Output Shape:", model.output_shape)
```

```python
import os

# Get all .jpg files from your dataset folder
image_files = [img for img in os.listdir(images_path) if img.lower().endswith('.jpg')]

print("Total images found:", len(image_files))
```

```python
def extract_features(images_list):
    features = {}

    for img_name in tqdm(images_list):
        img_path = os.path.join(images_path, img_name)

        try:
            # Load image and resize to 224x224 (required for VGG16)
            image = load_img(img_path, target_size=(224, 224))

            # Convert image to numpy array
            image = img_to_array(image)

            # Reshape image for model input: (1, 224, 224, 3)
            image = image.reshape((1, image.shape[0], image.shape[1], image.shape[2]))

            # Preprocess image for VGG16
            image = preprocess_input(image)

            # Extract feature vector
            feature = model.predict(image, verbose=0)

            # Store using image ID (without .jpg extension)
            image_id = img_name.split('.')[0]
            features[image_id] = feature

        except Exception as e:
            print(f"Error processing {img_name}: {e}")

    return features

print("Feature extraction function created successfully!")
```

```python
def clean(mapping):
    table = str.maketrans('', '', string.punctuation)

    for image_id, captions in mapping.items():
        for i in range(len(captions)):
            caption = captions[i]

            # lowercase
            caption = caption.lower()

            # remove punctuation
            caption = caption.translate(table)

            # split into words
            caption = caption.split()

            # remove short words
            caption = [word for word in caption if len(word) > 1]

            # keep only alphabetic words
            caption = [word for word in caption if word.isalpha()]

            # add start and end sequence tokens
            caption = 'startseq ' + ' '.join(caption) + ' endseq'

            captions[i] = caption

print("Caption cleaning function created successfully!")
```

```python
import os
import pickle
import numpy as np

from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, LSTM, Embedding, Dropout, add
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.utils import to_categorical
```

```python
# Correct paths for YOUR project
features_path = "../model/features.pkl"
tokenizer_path = "../model/tokenizer.pkl"

# Load features
with open(features_path, 'rb') as f:
    features = pickle.load(f)

# Load tokenizer
with open(tokenizer_path, 'rb') as f:
    tokenizer = pickle.load(f)

print("Features loaded successfully!")
print("Total feature vectors:", len(features))

print("Tokenizer loaded successfully!")
print("Vocabulary Size:", len(tokenizer.word_index) + 1)
```

```python
vocab_size = len(tokenizer.word_index) + 1
max_length = 34   # IMPORTANT: must match training

print("Vocabulary Size:", vocab_size)
print("Maximum Caption Length:", max_length)
```

```python
train_file = "../dataset/Flickr8k_text/Flickr_8k.trainImages.txt"

with open(train_file, 'r') as file:
    train_images = file.read().strip().split('\n')

# Remove .jpg extension
train_ids = [img.split('.')[0] for img in train_images]

print("Training image IDs loaded successfully!")
print("Total training images:", len(train_ids))
```

```python
train_mapping = {}

for image_id in train_ids:
    if image_id in mapping:
        train_mapping[image_id] = mapping[image_id]

print("Training captions mapping created successfully!")
print("Total training caption entries:", len(train_mapping))
```

```python
def data_generator(data_keys, mapping, features, tokenizer, max_length, vocab_size, batch_size):
    X1, X2, y = [], [], []
    n = 0

    while True:
        for key in data_keys:
            n += 1

            captions = mapping[key]

            for caption in captions:
                # Convert caption to integer sequence
                seq = tokenizer.texts_to_sequences([caption])[0]

                # Create multiple input-output pairs
                for i in range(1, len(seq)):
                    in_seq = seq[:i]
                    out_seq = seq[i]

                    # Pad input sequence
                    in_seq = pad_sequences([in_seq], maxlen=max_length)[0]

                    # One-hot encode output word
                    out_seq = to_categorical([out_seq], num_classes=vocab_size)[0]

                    # Image feature: (1,4096) -> (4096,)
                    img_feature = features[key][0]

                    # Append to batch
                    X1.append(img_feature)
                    X2.append(in_seq)
                    y.append(out_seq)

            # Yield batch
            if n == batch_size:
                yield (np.array(X1), np.array(X2)), np.array(y)
                X1, X2, y = [], [], []
                n = 0

print("Data generator created successfully!")
```

```python
# -----------------------------
# Image feature extractor branch
# -----------------------------
inputs1 = Input(shape=(4096,))
fe1 = Dropout(0.4)(inputs1)
fe2 = Dense(256, activation='relu')(fe1)

# -----------------------------
# Sequence model branch
# -----------------------------
inputs2 = Input(shape=(max_length,))
se1 = Embedding(vocab_size, 256, mask_zero=True)(inputs2)
se2 = Dropout(0.4)(se1)
se3 = LSTM(256)(se2)

# -----------------------------
# Decoder (merge both branches)
# -----------------------------
decoder1 = add([fe2, se3])
decoder2 = Dense(256, activation='relu')(decoder1)
outputs = Dense(vocab_size, activation='softmax')(decoder2)

# -----------------------------
# Final model
# -----------------------------
model = Model(inputs=[inputs1, inputs2], outputs=outputs)

# Compile model
model.compile(loss='categorical_crossentropy', optimizer='adam')

print("LSTM-based caption model built successfully!")
model.summary()
```

```python
# Training settings
epochs = 20
batch_size = 32
steps = len(train_ids) // batch_size

# Create generator
generator = data_generator(
    train_ids,
    train_mapping,
    features,
    tokenizer,
    max_length,
    vocab_size,
    batch_size
)

# Train model
history = model.fit(
    generator,
    epochs=epochs,
    steps_per_epoch=steps,
    verbose=1
)

print("Model training completed successfully!")
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
import numpy as np

def idx_to_word(integer, tokenizer):
    for word, index in tokenizer.word_index.items():
        if index == integer:
            return word
    return None


def predict_caption_greedy(model, image_feature, tokenizer, max_length):
    in_text = 'startseq'

    for i in range(max_length):
        sequence = tokenizer.texts_to_sequences([in_text])[0]
        sequence = pad_sequences([sequence], maxlen=max_length)

        yhat = model.predict([image_feature, sequence], verbose=0)
        yhat = np.argmax(yhat)

        word = idx_to_word(yhat, tokenizer)

        if word is None:
            break

        in_text += ' ' + word

        if word == 'endseq':
            break

    final_caption = in_text.replace('startseq', '').replace('endseq', '').strip()
    return final_caption

print("Greedy caption generation function created successfully!")
```

```python
import cv2
import torch
import warnings
from PIL import Image
from transformers import BlipProcessor, BlipForConditionalGeneration
from ultralytics import YOLO
from fer import FER

warnings.filterwarnings("ignore")

print("VISION AI GOD MODE STARTING...")

# -----------------------------
# DEVICE
# -----------------------------
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)

# -----------------------------
# LOAD BLIP CAPTION MODEL
# -----------------------------
processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
caption_model = BlipForConditionalGeneration.from_pretrained(
    "Salesforce/blip-image-captioning-base"
).to(device)

print("Caption AI Ready")

# -----------------------------
# LOAD YOLO OBJECT DETECTION
# -----------------------------
yolo_model = YOLO("yolov8n.pt")
print("Object Detection Ready")

# -----------------------------
# LOAD EMOTION DETECTOR
# -----------------------------
emotion_detector = FER(mtcnn=True)
print("Emotion Detection Ready")

# -----------------------------
# OPEN CAMERA
# -----------------------------
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    raise RuntimeError("Could not open webcam")

frame_count = 0
caption = "Starting..."
caption_interval = 60   # generate caption every 60 frames

print("\nPress 'q' in the webcam window to exit.\n")

try:
    while True:
        ret, frame = cap.read()
        if not ret:
            print("Failed to capture frame")
            break

        frame_count += 1

        # ---------------------------------
        # GENERATE CAPTION EVERY N FRAMES
        # ---------------------------------
        if frame_count % caption_interval == 0:
            try:
                rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
                image = Image.fromarray(rgb_frame)

                inputs = processor(images=image, return_tensors="pt").to(device)
                out = caption_model.generate(**inputs, max_new_tokens=25)

                caption = processor.decode(out[0], skip_special_tokens=True)
                print("Scene:", caption)

            except Exception as e:
                print("Caption Error:", e)

        # ---------------------------------
        # YOLO OBJECT DETECTION
        # ---------------------------------
        try:
            results = yolo_model(frame, verbose=False)
            annotated = results[0].plot()
        except Exception as e:
            print("YOLO Error:", e)
            annotated = frame.copy()

        # ---------------------------------
        # EMOTION DETECTION
        # ---------------------------------
        try:
            emotions = emotion_detector.detect_emotions(frame)

            for face in emotions:
                x, y, w, h = face["box"]
                emotion = max(face["emotions"], key=face["emotions"].get)

                # Draw face box
                cv2.rectangle(annotated, (x, y), (x + w, y + h), (255, 0, 0), 2)

                # Draw emotion label
                cv2.putText(
                    annotated,
                    f"Emotion: {emotion}",
                    (x, y - 10),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.7,
                    (255, 0, 0),
                    2
                )

        except Exception as e:
            # Sometimes FER may fail on some frames
            pass

        # ---------------------------------
        # DRAW CAPTION ON TOP
        # ---------------------------------
        # Background rectangle for caption
        cv2.rectangle(annotated, (10, 10), (annotated.shape[1] - 10, 60), (0, 0, 0), -1)

        # Limit long caption text
        display_caption = caption[:90]

        cv2.putText(
            annotated,
            f"Scene: {display_caption}",
            (20, 45),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.7,
            (0, 255, 0),
            2
        )

        # ---------------------------------
        # SHOW WINDOW
        # ---------------------------------
        cv2.imshow("VISION AI GOD MODE", annotated)

        # ---------------------------------
        # PRESS Q TO EXIT
        # ---------------------------------
        if cv2.waitKey(1) & 0xFF == ord("q"):
            print("Exiting webcam...")
            break

except KeyboardInterrupt:
    print("Stopped manually.")

finally:
    cap.release()
    cv2.destroyAllWindows()
    print("Webcam released successfully.")
```

```python
def generate_caption_blip(image_path, max_length=35, num_beams=5):
    """
    Generate caption using BLIP
    """
    if not os.path.exists(image_path):
        raise FileNotFoundError(f"❌ Image not found: {image_path}")

    image = PILImage.open(image_path).convert("RGB")

    inputs = processor(images=image, return_tensors="pt").to(device)

    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_length=max_length,
            num_beams=num_beams,
            early_stopping=True
        )

    caption = processor.decode(output[0], skip_special_tokens=True)
    return caption.strip().capitalize()
```

```python
import os
import io
import pickle
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

import nltk
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction

# TensorFlow / Keras
from tensorflow.keras.models import load_model, Model
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.preprocessing.image import img_to_array
from tensorflow.keras.applications.vgg16 import VGG16, preprocess_input

# BLIP
import torch
import PIL.Image as PILImage
from transformers import BlipProcessor, BlipForConditionalGeneration
```

```python
def load_flickr8k_captions(captions_file_path):
    mapping = {}

    if not os.path.exists(captions_file_path):
        print("⚠️ Flickr8k caption file not found. Ground-truth captions disabled.")
        return mapping

    with open(captions_file_path, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue

            parts = line.split("\t")
            if len(parts) != 2:
                continue

            image_caption_id, caption = parts
            image_id = image_caption_id.split("#")[0]

            if image_id not in mapping:
                mapping[image_id] = []

            mapping[image_id].append(caption)

    return mapping

captions_map = load_flickr8k_captions(captions_file)
print(f"✅ Loaded captions for {len(captions_map)} images")
```

```python
cnn_available = True
caption_model = None
tokenizer = None
feature_extractor = None
max_length = 34

try:
    # Load tokenizer
    with open(tokenizer_path, "rb") as f:
        tokenizer = pickle.load(f)

    # Load trained caption model
    caption_model = load_model(cnn_model_path)

    # Load VGG16 feature extractor
    base_model = VGG16()
    feature_extractor = Model(
        inputs=base_model.inputs,
        outputs=base_model.layers[-2].output
    )

    # Reverse mapping
    index_to_word = {index: word for word, index in tokenizer.word_index.items()}

    print("✅ CNN+LSTM model loaded successfully")

except Exception as e:
    cnn_available = False
    print("⚠️ CNN+LSTM model not available. UI will continue with BLIP only.")
    print("Reason:", e)
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)

processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-large")
blip_model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-large").to(device)

print("✅ BLIP-Large model loaded successfully")
```

```python
def idx_to_word(integer):
    if not cnn_available:
        return None
    return index_to_word.get(integer, None)


def extract_feature_from_new_image(image_path):
    """
    Extract 4096-dim VGG16 feature from image
    """
    image = Image.open(image_path).convert("RGB")
    image = image.resize((224, 224))
    image = img_to_array(image)
    image = np.expand_dims(image, axis=0)
    image = preprocess_input(image)

    feature = feature_extractor.predict(image, verbose=0)
    return feature[0]


def compute_bleu_score_multiple_references(references, generated_caption):
    """
    Compute BLEU score using multiple references
    """
    if not references or not generated_caption.strip():
        return None

    references_tokens = [ref.lower().split() for ref in references]
    candidate_tokens = generated_caption.lower().split()

    smoothie = SmoothingFunction().method1

    bleu = sentence_bleu(
        references_tokens,
        candidate_tokens,
        smoothing_function=smoothie
    )

    return round(bleu, 4)
```

```python
def generate_caption_beam_search_fixed(image_feature, beam_index=5, max_repeat=2):
    """
    Improved beam search decoding with repetition control.
    """
    if not cnn_available:
        return "CNN+LSTM model not available"

    start_token = tokenizer.word_index.get("startseq")
    end_token = tokenizer.word_index.get("endseq")

    if start_token is None or end_token is None:
        return "Tokenizer missing startseq/endseq"

    sequences = [[[start_token], 0.0]]

    for _ in range(max_length):
        all_candidates = []

        for seq, score in sequences:
            if seq[-1] == end_token:
                all_candidates.append((seq, score))
                continue

            padded_seq = pad_sequences([seq], maxlen=max_length)
            yhat = caption_model.predict([np.array([image_feature]), padded_seq], verbose=0)[0]

            top_indices = np.argsort(yhat)[-beam_index:]

            for idx in top_indices:
                prob = max(yhat[idx], 1e-10)
                candidate_seq = seq + [idx]
                candidate_score = score - np.log(prob)
                all_candidates.append((candidate_seq, candidate_score))

        ordered = sorted(all_candidates, key=lambda tup: tup[1])
        sequences = ordered[:beam_index]

    best_seq = sequences[0][0]

    caption_words = []
    word_counts = {}

    for idx in best_seq:
        word = idx_to_word(idx)

        if word is None:
            continue
        if word == "startseq":
            continue
        if word == "endseq":
            break

        # repetition control
        if word_counts.get(word, 0) >= max_repeat:
            continue
        if len(caption_words) > 0 and caption_words[-1] == word:
            continue

        # limit stop-word repetition
        if word in ["in", "on", "the", "a", "and"] and word_counts.get(word, 0) >= 2:
            continue

        caption_words.append(word)
        word_counts[word] = word_counts.get(word, 0) + 1

    final_caption = " ".join(caption_words).strip()

    if not final_caption:
        final_caption = "Could not generate a good caption"

    return final_caption
```

```python
def generate_caption_blip(image_path, prompt="a photography of", max_length=40, num_beams=7):
    """
    BLIP-Large caption generation with prompt guidance
    """
    image = PILImage.open(image_path).convert("RGB")

    inputs = processor(images=image, text=prompt, return_tensors="pt").to(device)

    with torch.no_grad():
        output = blip_model.generate(
            **inputs,
            max_length=max_length,
            num_beams=num_beams,
            early_stopping=True
        )

    caption = processor.decode(output[0], skip_special_tokens=True)
    return caption.strip()
```

```python
import torch
from PIL import Image
import matplotlib.pyplot as plt
from transformers import BlipProcessor, BlipForConditionalGeneration
```

```python
# Device setup
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)

# Load BLIP model and processor
processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-base").to(device)

print("BLIP model loaded successfully!")
```

```python
def generate_caption(image, prompt=None, max_length=40, num_beams=5):
    """
    Generate caption using BLIP with optional text prompt.
    """
    if prompt:
        inputs = processor(images=image, text=prompt, return_tensors="pt").to(device)
    else:
        inputs = processor(images=image, return_tensors="pt").to(device)

    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_length=max_length,
            num_beams=num_beams,
            early_stopping=True
        )

    caption = processor.decode(output[0], skip_special_tokens=True)
    return caption.strip().capitalize()


def multi_perspective_captioning(image_path, show_image=True):
    """
    Generate multiple captions for the same image from different perspectives.
    """
    image = Image.open(image_path).convert("RGB")

    if show_image:
        plt.figure(figsize=(8, 6))
        plt.imshow(image)
        plt.axis("off")
        plt.title("Input Image")
        plt.show()

    # Different perspectives using prompt engineering
    perspectives = {
        "1. General Caption": None,
        "2. Object-Focused Caption": "Describe the main objects in this image",
        "3. Scene-Focused Caption": "Describe the overall scene and environment in this image",
        "4. Detailed Caption": "Provide a detailed description of this image",
        "5. Short Summary Caption": "Give a short caption for this image"
    }

    results = {}

    for perspective_name, prompt in perspectives.items():
        try:
            caption = generate_caption(image, prompt=prompt)
            results[perspective_name] = caption
        except Exception as e:
            results[perspective_name] = f"Error: {str(e)}"

    return results
```

```python
import torch
from PIL import Image
import matplotlib.pyplot as plt

from transformers import BlipProcessor, BlipForConditionalGeneration
from torchvision import models, transforms
from torchvision.models import ResNet50_Weights
```

```python
# Device setup
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)

# -----------------------------
# 1. BLIP model for captioning
# -----------------------------
blip_processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
blip_model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-base").to(device)

# ----------------------------------------
# 2. ResNet50 model for scene clue analysis
# ----------------------------------------
resnet_weights = ResNet50_Weights.DEFAULT
resnet_model = models.resnet50(weights=resnet_weights).to(device)
resnet_model.eval()

# Preprocessing transform for ResNet
resnet_preprocess = resnet_weights.transforms()

# Class labels from torchvision weights meta
imagenet_labels = resnet_weights.meta["categories"]

print("Models loaded successfully!")
```

```python
# Keywords that loosely indicate positive / negative mood
happy_keywords = [
    "beach", "seashore", "sunrise", "sunset", "park", "garden", "flower",
    "daisy", "sunflower", "balloon", "kite", "ice cream", "toyshop",
    "golden retriever", "labrador retriever", "puppy", "tabby", "Persian cat",
    "soccer ball", "basketball", "groom", "bride", "stage", "fountain"
]

sad_keywords = [
    "umbrella", "rain barrel", "storm", "wreck", "chainlink fence",
    "prison", "streetlight", "ashcan", "garbage truck", "traffic light",
    "window screen", "torch", "dock", "pier", "lifeboat"
]

# Optional contextual words you can inject
emotion_prefix = {
    "happy": [
        "A cheerful scene of",
        "A joyful view of",
        "A pleasant and lively image showing"
    ],
    "sad": [
        "A somber scene of",
        "A quiet and melancholic image of",
        "A gloomy view of"
    ],
    "neutral": [
        "An image of",
        "A scene showing",
        "A picture of"
    ]
}
```

```python
import random
import re

def clean_caption(text):
    """
    Basic cleanup for generated caption text.
    """
    text = text.strip()
    text = re.sub(r"\s+", " ", text)
    text = text[0].upper() + text[1:] if len(text) > 1 else text
    return text

def generate_base_caption(image):
    """
    Generate a base caption using BLIP.
    """
    inputs = blip_processor(images=image, return_tensors="pt").to(device)

    with torch.no_grad():
        output_ids = blip_model.generate(
            **inputs,
            max_length=30,
            num_beams=5,
            early_stopping=True
        )

    caption = blip_processor.decode(output_ids[0], skip_special_tokens=True)
    return clean_caption(caption)

def predict_top_labels(image, topk=5):
    """
    Predict top-k ImageNet labels using ResNet50.
    """
    img_tensor = resnet_preprocess(image).unsqueeze(0).to(device)

    with torch.no_grad():
        logits = resnet_model(img_tensor)
        probs = torch.nn.functional.softmax(logits, dim=1)

    top_probs, top_idxs = torch.topk(probs, topk)
    results = []
    for p, idx in zip(top_probs[0], top_idxs[0]):
        label = imagenet_labels[idx.item()]
        results.append((label.lower(), float(p.item())))
    return results

def detect_emotion_from_labels(top_labels):
    """
    Rule-based emotion detection from top predicted labels.
    """
    labels_only = [label for label, _ in top_labels]

    happy_score = 0
    sad_score = 0

    for label in labels_only:
        for kw in happy_keywords:
            if kw in label:
                happy_score += 1
        for kw in sad_keywords:
            if kw in label:
                sad_score += 1

    # Heuristic decision
    if happy_score > sad_score and happy_score >= 1:
        return "happy"
    elif sad_score > happy_score and sad_score >= 1:
        return "sad"
    else:
        return "neutral"

def make_emotion_aware_caption(base_caption, emotion):
    """
    Convert the base caption into an emotion-aware caption.
    """
    prefix = random.choice(emotion_prefix[emotion])

    # Remove leading "a", "an", "the" for smoother rewriting
    lower_caption = base_caption.lower()
    for article in ["a ", "an ", "the "]:
        if lower_caption.startswith(article):
            base_caption = base_caption[len(article):]
            break

    # Emotion-specific enhancement
    if emotion == "happy":
        return f"{prefix} {base_caption.lower()}, creating a warm and uplifting atmosphere."
    elif emotion == "sad":
        return f"{prefix} {base_caption.lower()}, giving the image a subdued and emotional tone."
    else:
        return f"{prefix} {base_caption.lower()}."

def emotion_aware_caption_pipeline(image_path, show_image=True):
    """
    Full pipeline:
    1. Load image
    2. Generate base caption
    3. Predict scene labels
    4. Detect emotion
    5. Generate emotion-aware caption
    """
    image = Image.open(image_path).convert("RGB")

    if show_image:
        plt.figure(figsize=(8, 6))
        plt.imshow(image)
        plt.axis("off")
        plt.title("Input Image")
        plt.show()

    # Step 1: Base caption
    base_caption = generate_base_caption(image)

    # Step 2: Scene clues
    top_labels = predict_top_labels(image, topk=5)

    # Step 3: Emotion
    emotion = detect_emotion_from_labels(top_labels)

    # Step 4: Emotion-aware caption
    final_caption = make_emotion_aware_caption(base_caption, emotion)

    return {
        "base_caption": base_caption,
        "top_labels": top_labels,
        "predicted_emotion": emotion,
        "emotion_aware_caption": final_caption
    }
```

```python
image_path = r"C:\Users\SAIF CHIKTE\Image_captioning\dataset\Flicker8k_Dataset\57417274_d55d34e93e.jpg"  # Change this to your image path

result = emotion_aware_caption_pipeline(image_path)

print("Base Caption:")
print(result["base_caption"])

print("\nTop Scene Labels:")
for label, score in result["top_labels"]:
    print(f"- {label}: {score:.4f}")

print("\nPredicted Emotion:")
print(result["predicted_emotion"])

print("\nEmotion-Aware Caption:")
print(result["emotion_aware_caption"])
```

```python
import torch
from transformers import BlipProcessor, BlipForConditionalGeneration
from PIL import Image
import matplotlib.pyplot as plt
import os
```

```python
# Load BLIP processor + model
processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
blip_model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-base")

# Use GPU if available
device = "cuda" if torch.cuda.is_available() else "cpu"
blip_model.to(device)

print(f"✅ BLIP model loaded on {device}")
```

```python
def generate_blip_caption(image_path):
    # Open image
    raw_image = Image.open(image_path).convert("RGB")

    # Process image
    inputs = processor(images=raw_image, return_tensors="pt").to(device)

    # Generate caption
    out = blip_model.generate(**inputs, max_new_tokens=30)

    # Decode
    caption = processor.decode(out[0], skip_special_tokens=True)

    return caption
```

```python
def personalize_caption(base_caption, style="default", user_name=None):
    caption = base_caption.strip()

    # Capitalize first letter
    if caption:
        caption = caption[0].upper() + caption[1:]

    if style == "simple":
        words = caption.split()
        caption = " ".join(words[:8]) if len(words) > 8 else caption

    elif style == "detailed":
        caption = f"This image appears to show {caption.lower()}, with clear visual details in the scene."

    elif style == "social_media":
        caption = f"✨ {caption}!"

    elif style == "child_friendly":
        caption = f"This picture shows {caption.lower()}. It looks fun and exciting!"

    elif style == "formal":
        caption = f"The image depicts {caption.lower()}."

    return caption
```

```python
def generate_personalized_caption_blip(image_path, style="default", user_name=None, show_image=True):
    # Check file exists
    if not os.path.exists(image_path):
        raise FileNotFoundError(f"❌ Image not found: {image_path}")

    # Show image
    if show_image:
        img = Image.open(image_path).convert("RGB")
        plt.figure(figsize=(8, 6))
        plt.imshow(img)
        plt.axis("off")
        plt.title("Input Image")
        plt.show()

    # Generate base caption using BLIP
    base_caption = generate_blip_caption(image_path)

    # Personalize
    personalized = personalize_caption(base_caption, style=style, user_name=user_name)

    print("📝 Base Caption:")
    print(base_caption)

    print("\n🎯 Personalized Caption:")
    print(personalized)

    return base_caption, personalized
```

```python
def load_all_preferences():
    with open(PREFERENCES_FILE, "r") as f:
        return json.load(f)
        
```

```python
def save_all_preferences(preferences):
    with open(PREFERENCES_FILE, "w") as f:
        json.dump(preferences, f, indent=4)
```

```python
def update_user_preference(user_name, style="default", language="English", tone="neutral"):
    preferences = load_all_preferences()

    if user_name not in preferences:
        preferences[user_name] = {
            "preferred_style": style,
            "preferred_language": language,
            "preferred_tone": tone,
            "usage_count": 1
        }
    else:
        preferences[user_name]["preferred_style"] = style
        preferences[user_name]["preferred_language"] = language
        preferences[user_name]["preferred_tone"] = tone
        preferences[user_name]["usage_count"] += 1

    save_all_preferences(preferences)

    # Clean output (no user name shown)
    print("✅ Preference learned successfully : ")
    print(preferences[user_name])
```

```python
def get_user_preference(user_name):
    preferences = load_all_preferences()

    if user_name in preferences:
        return preferences[user_name]
    else:
        return {
            "preferred_style": "default",
            "preferred_language": "English",
            "preferred_tone": "neutral",
            "usage_count": 0
        }
```

```python
def personalize_caption(base_caption, style="default", user_name=None):
    caption = base_caption.strip()

    # Capitalize first letter
    if caption:
        caption = caption[0].upper() + caption[1:]

    if style == "simple":
        words = caption.split()
        caption = " ".join(words[:8]) if len(words) > 8 else caption

    elif style == "detailed":
        caption = f"This image appears to show {caption.lower()}, with clear visual details in the scene."

    elif style == "social_media":
        caption = f"✨ {caption}! "

    elif style == "child_friendly":
        caption = f"This picture shows {caption.lower()}. It looks fun and exciting!"

    elif style == "formal":
        caption = f"The image depicts {caption.lower()}."

    # Removed user name prefix
    return caption
```

```python
def generate_caption_with_learned_preferences(image_path, user_name, style=None, language=None, tone=None, show_image=True):
    # Check image path
    if not os.path.exists(image_path):
        raise FileNotFoundError(f"❌ Image not found: {image_path}")

    # Load old preferences
    old_pref = get_user_preference(user_name)

    # If user provides new preference, learn it
    final_style = style if style is not None else old_pref["preferred_style"]
    final_language = language if language is not None else old_pref["preferred_language"]
    final_tone = tone if tone is not None else old_pref["preferred_tone"]

    # Save updated preference (learning)
    update_user_preference(user_name, final_style, final_language, final_tone)

    # Show image
    if show_image:
        img = Image.open(image_path).convert("RGB")
        plt.figure(figsize=(8, 6))
        plt.imshow(img)
        plt.axis("off")
        plt.title("Input Image")
        plt.show()

    # Generate base caption using BLIP
    base_caption = generate_blip_caption(image_path)

    # Apply personalized caption using learned style
    personalized_caption = personalize_caption(base_caption, style=final_style, user_name=user_name)

    # Display results
    print("👤 Learned User Preference:")
    print(get_user_preference(user_name))

    print("\n📝 Base Caption:")
    print(base_caption)

    print("\n🎯 Personalized Caption (Using Learned Preference):")
    print(personalized_caption)

    return base_caption, personalized_caption
```

```python
import os
import torch
import matplotlib.pyplot as plt
from PIL import Image
from transformers import BlipProcessor, BlipForConditionalGeneration
```

```python
# Load BLIP processor + model
processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
blip_model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-base")

# Device setup
device = "cuda" if torch.cuda.is_available() else "cpu"
blip_model.to(device)

print(f"✅ BLIP model loaded on {device}")
```

```python
def generate_blip_caption(image_path):
    raw_image = Image.open(image_path).convert("RGB")
    inputs = processor(images=raw_image, return_tensors="pt").to(device)

    output = blip_model.generate(**inputs, max_new_tokens=30)
    caption = processor.decode(output[0], skip_special_tokens=True)

    return caption
```

```python
def clean_caption_text(caption):
    caption = caption.strip()
    if caption:
        caption = caption[0].upper() + caption[1:]
    return caption
```

```python
def create_narrative_story(captions):
    """
    Simple rule-based story generation from multiple image captions.
    """
    cleaned_captions = [clean_caption_text(c) for c in captions if c.strip()]

    if len(cleaned_captions) == 0:
        return "No meaningful story could be generated from the provided images."

    if len(cleaned_captions) == 1:
        return f"The story begins with {cleaned_captions[0].lower()}."

    # Start story
    story = "A visual story unfolds. "

    # First image
    story += f"First, we see {cleaned_captions[0].lower()}. "

    # Middle images
    for i in range(1, len(cleaned_captions) - 1):
        story += f"Then, the scene continues with {cleaned_captions[i].lower()}. "

    # Last image
    story += f"Finally, the story concludes with {cleaned_captions[-1].lower()}."

    return story
```

```python
def generate_story_from_images(image_paths, show_images=True):
    # Validate input
    if not image_paths or len(image_paths) == 0:
        raise ValueError("❌ Please provide at least one image path.")

    captions = []

    # Show images
    if show_images:
        plt.figure(figsize=(5 * len(image_paths), 5))

    for idx, image_path in enumerate(image_paths):
        if not os.path.exists(image_path):
            raise FileNotFoundError(f"❌ Image not found: {image_path}")

        # Generate caption
        caption = generate_blip_caption(image_path)
        captions.append(caption)

        # Display image
        if show_images:
            img = Image.open(image_path).convert("RGB")
            plt.subplot(1, len(image_paths), idx + 1)
            plt.imshow(img)
            plt.axis("off")
            plt.title(f"Image {idx+1}")

    if show_images:
        plt.tight_layout()
        plt.show()

    # Print individual captions
    print("📝 Individual Image Captions:")
    for i, cap in enumerate(captions, start=1):
        print(f"Image {i}: {cap}")

    # Generate final story
    story = create_narrative_story(captions)

    print("\n📖 Generated Narrative Story:")
    print(story)

    return captions, story
```

```python
image_list = [
    r"C:\Users\SAIF CHIKTE\Image_captioning\dataset\Flicker8k_Dataset\55470226_52ff517151.jpg",
    r"C:\Users\SAIF CHIKTE\Image_captioning\dataset\Flicker8k_Dataset\96399948_b86c61bfe6.jpg",
    r"C:\Users\SAIF CHIKTE\Image_captioning\dataset\Flicker8k_Dataset\111537217_082a4ba060.jpg"
]

captions, story = generate_story_from_images(image_list)
```

```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import cv2
from PIL import Image
```

```python
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.applications.resnet50 import preprocess_input
from tensorflow.keras.models import Model

# Load pretrained ResNet50
base_model = ResNet50(weights='imagenet')

print("✅ ResNet50 loaded successfully")
```

```python
def load_and_preprocess_image(image_path, target_size=(224, 224)):
    # Load image using PIL
    img = Image.open(image_path).convert("RGB")
    img_resized = img.resize(target_size)

    # Convert to numpy
    img_array = np.array(img_resized)
    img_batch = np.expand_dims(img_array, axis=0)
    img_batch = preprocess_input(img_batch.astype(np.float32))

    return img, img_array, img_batch
```

```python
def make_gradcam_heatmap(img_batch, model, last_conv_layer_name="conv5_block3_out"):
    # Create a model that outputs:
    # 1. last conv layer activations
    # 2. final predictions
    grad_model = Model(
        inputs=model.inputs,
        outputs=[model.get_layer(last_conv_layer_name).output, model.output]
    )

    with tf.GradientTape() as tape:
        conv_outputs, predictions = grad_model(img_batch)
        pred_index = tf.argmax(predictions[0])
        class_channel = predictions[:, pred_index]

    # Compute gradients of predicted class wrt conv outputs
    grads = tape.gradient(class_channel, conv_outputs)

    # Global average pooling of gradients
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))

    # Weight conv outputs
    conv_outputs = conv_outputs[0]
    heatmap = conv_outputs @ pooled_grads[..., tf.newaxis]
    heatmap = tf.squeeze(heatmap)

    # Normalize heatmap
    heatmap = np.maximum(heatmap, 0)
    if np.max(heatmap) != 0:
        heatmap /= np.max(heatmap)

    return heatmap
```

```python
def overlay_heatmap_on_image(original_img_array, heatmap, alpha=0.4):
    # Resize heatmap to match image size
    heatmap_resized = cv2.resize(heatmap, (original_img_array.shape[1], original_img_array.shape[0]))

    # Convert heatmap to color map
    heatmap_uint8 = np.uint8(255 * heatmap_resized)
    heatmap_color = cv2.applyColorMap(heatmap_uint8, cv2.COLORMAP_JET)

    # Overlay
    original_uint8 = original_img_array.astype(np.uint8)
    overlay = cv2.addWeighted(original_uint8, 1 - alpha, heatmap_color, alpha, 0)

    return heatmap_resized, overlay
```

```python
def visualize_attention_mechanism(image_path):
    # Step 1: Load image
    _, original_img_array, img_batch = load_and_preprocess_image(image_path)

    # Step 2: Generate heatmap
    heatmap = make_gradcam_heatmap(
        img_batch=img_batch,
        model=base_model,
        last_conv_layer_name="conv5_block3_out"
    )

    # Step 3: Create overlay
    heatmap_resized, overlay = overlay_heatmap_on_image(original_img_array, heatmap)

    # Step 4: Show results
    plt.figure(figsize=(18, 6))

    # Original Image
    plt.subplot(1, 3, 1)
    plt.imshow(original_img_array)
    plt.title("Original Image")
    plt.axis("off")

    # Heatmap
    plt.subplot(1, 3, 2)
    plt.imshow(heatmap_resized, cmap="jet")
    plt.title("Attention Heatmap")
    plt.axis("off")

    # Overlay
    plt.subplot(1, 3, 3)
    plt.imshow(cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB))
    plt.title("Attention Overlay")
    plt.axis("off")

    plt.tight_layout()
    plt.show()

    print("✅ Attention heatmap generated successfully.")
    print("The highlighted regions indicate where the model focuses most strongly in the image.")

    return heatmap
```

```python
test_image_path = r"C:\Users\SAIF CHIKTE\Image_captioning\dataset\Flicker8k_Dataset\47870024_73a4481f7d.jpg"

heatmap = visualize_attention_mechanism(test_image_path)
```

```python
!pip install ultralytics transformers torch torchvision sentencepiece accelerate opencv-python matplotlib pillow
```

```python
import os
import cv2
import torch
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

from ultralytics import YOLO
from transformers import BlipProcessor, BlipForConditionalGeneration
```

```python
# Load YOLO model (small and fast)
yolo_model = YOLO("yolov8n.pt")

# Load BLIP caption model
processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
blip_model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-base")

# Device setup
device = "cuda" if torch.cuda.is_available() else "cpu"
blip_model.to(device)

print(f"✅ YOLO loaded")
print(f"✅ BLIP loaded on {device}")
```

```python
def generate_region_caption(pil_image):
    # Ensure RGB
    region_image = pil_image.convert("RGB")

    # Process
    inputs = processor(images=region_image, return_tensors="pt").to(device)

    # Generate caption
    output = blip_model.generate(**inputs, max_new_tokens=20)
    caption = processor.decode(output[0], skip_special_tokens=True)

    return caption
```

```python
def dense_region_level_captioning(image_path, conf_threshold=0.25, show_image=True):
    if not os.path.exists(image_path):
        raise FileNotFoundError(f"❌ Image not found: {image_path}")

    # Load image
    image_bgr = cv2.imread(image_path)
    image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
    pil_image = Image.fromarray(image_rgb)

    # Run YOLO detection
    results = yolo_model(image_path, conf=conf_threshold, verbose=False)

    detections = []
    annotated_image = image_rgb.copy()

    for result in results:
        boxes = result.boxes

        if boxes is None:
            continue

        for box in boxes:
            # Get coordinates
            x1, y1, x2, y2 = map(int, box.xyxy[0].tolist())
            conf = float(box.conf[0])
            cls_id = int(box.cls[0])
            class_name = yolo_model.names[cls_id]

            # Safety clamp
            x1 = max(0, x1)
            y1 = max(0, y1)
            x2 = min(image_rgb.shape[1], x2)
            y2 = min(image_rgb.shape[0], y2)

            # Skip invalid boxes
            if x2 <= x1 or y2 <= y1:
                continue

            # Crop region
            crop = image_rgb[y1:y2, x1:x2]
            crop_pil = Image.fromarray(crop)

            # Generate caption for region
            region_caption = generate_region_caption(crop_pil)

            # Save result
            detections.append({
                "class_name": class_name,
                "confidence": conf,
                "bbox": (x1, y1, x2, y2),
                "caption": region_caption
            })

            # Draw bounding box
            cv2.rectangle(annotated_image, (x1, y1), (x2, y2), (0, 255, 0), 2)

            # Label text
            label = f"{class_name}"
            cv2.putText(
                annotated_image,
                label,
                (x1, max(y1 - 10, 20)),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.6,
                (0, 255, 0),
                2
            )

    # Show annotated image
    if show_image:
        plt.figure(figsize=(10, 8))
        plt.imshow(annotated_image)
        plt.axis("off")
        plt.title("Dense Region-Level Captioning (YOLO + BLIP)")
        plt.show()

    # Print region captions
    print("📝 Region-Level Captions:")
    if len(detections) == 0:
        print("No objects detected.")
    else:
        for i, det in enumerate(detections, start=1):
            print(f"\nRegion {i}:")
            print(f"  Object: {det['class_name']}")
            print(f"  Confidence: {det['confidence']:.2f}")
            print(f"  Bounding Box: {det['bbox']}")
            print(f"  Caption: {det['caption']}")

    return detections, annotated_image
```

```python
test_image_path = r"C:\Users\SAIF CHIKTE\Image_captioning\dataset\Flicker8k_Dataset\47870024_73a4481f7d.jpg"

detections, annotated_img = dense_region_level_captioning(
    image_path=test_image_path,
    conf_threshold=0.25
)
```

```python
import warnings
warnings.filterwarnings("ignore")

import torch
import cv2
import numpy as np
import matplotlib.pyplot as plt

from PIL import Image
from collections import Counter
from ultralytics import YOLO
from transformers import BlipProcessor, BlipForConditionalGeneration
```

```python
# Device setup
device = "cuda" if torch.cuda.is_available() else "cpu"

# BLIP model
processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
blip_model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-base").to(device)

# YOLOv8 model
yolo_model = YOLO("yolov8n.pt")

print("Models loaded successfully.")
print(f"Using device: {device}")
```

```python
def display_image(image, title="Image", figsize=(9, 6)):
    plt.figure(figsize=figsize)
    plt.imshow(image)
    plt.axis("off")
    plt.title(title)
    plt.show()
```

```python
def generate_blip_caption(pil_image, max_new_tokens=25):
    """
    Generate a clean BLIP caption for a PIL image.
    """
    inputs = processor(images=pil_image, return_tensors="pt").to(device)

    with torch.no_grad():
        output = blip_model.generate(**inputs, max_new_tokens=max_new_tokens)

    caption = processor.decode(output[0], skip_special_tokens=True).strip()
    return caption
```

```python
def compute_iou(boxA, boxB):
    """
    Compute Intersection over Union (IoU) between two boxes.
    Boxes format: (x1, y1, x2, y2)
    """
    xA = max(boxA[0], boxB[0])
    yA = max(boxA[1], boxB[1])
    xB = min(boxA[2], boxB[2])
    yB = min(boxA[3], boxB[3])

    inter_w = max(0, xB - xA)
    inter_h = max(0, yB - yA)
    inter_area = inter_w * inter_h

    boxA_area = max(0, boxA[2] - boxA[0]) * max(0, boxA[3] - boxA[1])
    boxB_area = max(0, boxB[2] - boxB[0]) * max(0, boxB[3] - boxB[1])

    union_area = boxA_area + boxB_area - inter_area + 1e-6

    return inter_area / union_area
```

```python
def filter_detections(
    result,
    allowed_classes=None,
    min_conf=0.55,
    min_crop_size=60,
    max_regions=5,
    iou_threshold=0.50
):
    """
    Filter YOLO detections for clean hierarchical captioning.

    Steps:
    - confidence filter
    - allowed class filter
    - minimum crop size filter
    - sort by confidence
    - remove duplicates using simple IoU suppression
    """
    filtered = []

    if result.boxes is None or len(result.boxes) == 0:
        return filtered

    for box in result.boxes:
        conf = float(box.conf[0].item())
        cls_id = int(box.cls[0].item())
        class_name = yolo_model.names[cls_id]

        if conf < min_conf:
            continue

        if allowed_classes is not None and class_name not in allowed_classes:
            continue

        x1, y1, x2, y2 = map(int, box.xyxy[0].tolist())
        w = max(0, x2 - x1)
        h = max(0, y2 - y1)

        if w < min_crop_size or h < min_crop_size:
            continue

        area = w * h

        filtered.append({
            "class_name": class_name,
            "confidence": conf,
            "bbox": (x1, y1, x2, y2),
            "area": area
        })

    # Sort by confidence first, then area
    filtered = sorted(filtered, key=lambda x: (x["confidence"], x["area"]), reverse=True)

    # Simple duplicate suppression using IoU
    final_detections = []
    for det in filtered:
        keep = True
        for kept in final_detections:
            # suppress only if same class and high overlap
            if det["class_name"] == kept["class_name"]:
                iou = compute_iou(det["bbox"], kept["bbox"])
                if iou > iou_threshold:
                    keep = False
                    break
        if keep:
            final_detections.append(det)

    return final_detections[:max_regions]
```

```python
def draw_filtered_boxes(image_np, detections):
    """
    Draw filtered bounding boxes manually for cleaner output.
    """
    annotated = image_np.copy()

    for det in detections:
        x1, y1, x2, y2 = det["bbox"]
        label = f"{det['class_name']} {det['confidence']:.2f}"

        # Box
        cv2.rectangle(annotated, (x1, y1), (x2, y2), (180, 0, 180), 2)

        # Label background
        (tw, th), _ = cv2.getTextSize(label, cv2.FONT_HERSHEY_SIMPLEX, 0.65, 2)
        cv2.rectangle(annotated, (x1, max(0, y1 - th - 10)), (x1 + tw + 8, y1), (180, 0, 180), -1)

        # Label text
        cv2.putText(
            annotated,
            label,
            (x1 + 4, y1 - 6),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.65,
            (255, 255, 255),
            2,
            cv2.LINE_AA
        )

    return annotated
```

```python
def generate_filtered_region_captions(
    image_path,
    allowed_classes=None,
    min_conf=0.55,
    min_crop_size=60,
    max_regions=5,
    iou_threshold=0.50
):
    """
    Detect objects using YOLO, apply practical filtering,
    then generate BLIP captions for selected regions.
    """
    pil_image = Image.open(image_path).convert("RGB")
    image_np = np.array(pil_image)

    # Run YOLO with clean output
    results = yolo_model.predict(source=image_np, verbose=False)
    result = results[0]

    # Filter detections
    detections = filter_detections(
        result=result,
        allowed_classes=allowed_classes,
        min_conf=min_conf,
        min_crop_size=min_crop_size,
        max_regions=max_regions,
        iou_threshold=iou_threshold
    )

    # Create clean annotated image from filtered boxes only
    annotated_rgb = draw_filtered_boxes(image_np, detections)

    region_data = []

    for idx, det in enumerate(detections, start=1):
        x1, y1, x2, y2 = det["bbox"]

        crop = image_np[max(0, y1):max(0, y2), max(0, x1):max(0, x2)]
        if crop.size == 0:
            continue

        crop_pil = Image.fromarray(crop)

        # Region caption
        region_caption = generate_blip_caption(crop_pil, max_new_tokens=20)

        region_data.append({
            "region_id": idx,
            "object": det["class_name"],
            "confidence": round(det["confidence"], 3),
            "bbox": det["bbox"],
            "caption": region_caption
        })

    return annotated_rgb, region_data
```

```python
def build_hierarchical_summary_v2(global_caption, region_data):
    """
    Build a cleaner hierarchical summary with:
    - global caption
    - object counts
    - concise region summary
    """
    if not region_data:
        return (
            f"Overall, the image shows {global_caption}. "
            f"No major object regions passed the filtering criteria, so the final description relies on the global scene understanding."
        )

    # Count objects
    object_counts = Counter([item["object"] for item in region_data])

    # Create object count phrase
    object_phrases = []
    for obj, count in object_counts.items():
        if count == 1:
            object_phrases.append(f"1 {obj}")
        else:
            object_phrases.append(f"{count} {obj}s")

    object_summary = ", ".join(object_phrases)

    # Region summary (top 3 concise)
    region_lines = []
    for item in region_data[:3]:
        obj = item["object"]
        cap = item["caption"]
        region_lines.append(f"{obj}: {cap}")

    region_summary = "; ".join(region_lines)

    final_summary = (
        f"Overall, the image shows {global_caption}. "
        f"At the object level, the system detected {object_summary}. "
        f"At the region level, the localized descriptions include {region_summary}. "
        f"This hierarchical caption combines global scene understanding with filtered object-aware regional details."
    )

    return final_summary
```

```python
def hierarchical_multi_level_caption_v2(
    image_path,
    allowed_classes=None,
    min_conf=0.55,
    min_crop_size=60,
    max_regions=5,
    iou_threshold=0.50
):
    """
    Final Clean v2:
    1. Global Caption
    2. Filtered Region-Level Captions
    3. Improved Hierarchical Summary

    Displays:
    - Original image
    - Filtered annotated image

    Returns:
    - results dictionary
    """
    # Load original image
    pil_image = Image.open(image_path).convert("RGB")

    # Show original image
    display_image(pil_image, title="Original Image")

    # Level 1: Global caption
    global_caption = generate_blip_caption(pil_image, max_new_tokens=30)

    # Level 2: Filtered region-level captions
    annotated_image, region_data = generate_filtered_region_captions(
        image_path=image_path,
        allowed_classes=allowed_classes,
        min_conf=min_conf,
        min_crop_size=min_crop_size,
        max_regions=max_regions,
        iou_threshold=iou_threshold
    )

    # Show filtered annotated image
    display_image(annotated_image, title="Filtered Detected Regions")

    # Level 3: Improved hierarchical summary
    hierarchical_summary = build_hierarchical_summary_v2(global_caption, region_data)

    # Clean professional output
    print("=" * 90)
    print("HIERARCHICAL MULTI-LEVEL CAPTION GENERATION (FILTERED FINAL v2)")
    print("=" * 90)

    print("\n[Level 1] Global Caption:")
    print(global_caption)

    print("\n[Level 2] Filtered Region-Level Captions:")
    if region_data:
        for item in region_data:
            print(f"Region {item['region_id']}: {item['object']} ({item['confidence']:.2f})")
            print(f"  Caption: {item['caption']}")
    else:
        print("No major regions passed the filtering criteria.")

    print("\n[Level 3] Hierarchical Scene Summary:")
    print(hierarchical_summary)

    print("\nFeature executed successfully.")

    return {
        "global_caption": global_caption,
        "region_captions": region_data,
        "hierarchical_summary": hierarchical_summary
    }
```

```python
import warnings
warnings.filterwarnings("ignore")

import torch
import pandas as pd
import matplotlib.pyplot as plt

from PIL import Image
from IPython.display import display
from transformers import BlipProcessor, BlipForConditionalGeneration
```

```python
# Device setup
device = "cuda" if torch.cuda.is_available() else "cpu"

# BLIP = Vision-Language Transformer
processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
vl_transformer_model = BlipForConditionalGeneration.from_pretrained(
    "Salesforce/blip-image-captioning-base"
).to(device)

print("Vision-Language Transformer (BLIP) loaded successfully.")
print(f"Using device: {device}")
```

```python
def display_image(image, title="Input Image", figsize=(8, 6)):
    plt.figure(figsize=figsize)
    plt.imshow(image)
    plt.axis("off")
    plt.title(title)
    plt.show()
```

```python
def generate_unconditional_caption(image, max_new_tokens=30):
    """
    Generate standard caption using BLIP (Vision-Language Transformer).
    """
    inputs = processor(images=image, return_tensors="pt").to(device)

    with torch.no_grad():
        output = vl_transformer_model.generate(
            **inputs,
            max_new_tokens=max_new_tokens
        )

    caption = processor.decode(output[0], skip_special_tokens=True).strip()
    return caption
```

```python
def generate_prompt_guided_caption(image, prompt="a photography of", max_new_tokens=30):
    """
    Generate prompt-guided caption to demonstrate multimodal conditioning.
    """
    inputs = processor(images=image, text=prompt, return_tensors="pt").to(device)

    with torch.no_grad():
        output = vl_transformer_model.generate(
            **inputs,
            max_new_tokens=max_new_tokens
        )

    caption = processor.decode(output[0], skip_special_tokens=True).strip()
    return caption
```

```python
def generate_beam_search_caption(image, max_new_tokens=30, num_beams=5):
    """
    Generate higher-quality caption using beam search.
    """
    inputs = processor(images=image, return_tensors="pt").to(device)

    with torch.no_grad():
        output = vl_transformer_model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            num_beams=num_beams,
            early_stopping=True
        )

    caption = processor.decode(output[0], skip_special_tokens=True).strip()
    return caption
```

```python
def create_caption_comparison_table(unconditional_caption, prompt_caption, beam_caption):
    data = {
        "Mode": [
            "Unconditional Caption",
            "Prompt-Guided Caption",
            "Beam Search Caption"
        ],
        "Generated Caption": [
            unconditional_caption,
            prompt_caption,
            beam_caption
        ]
    }
    return pd.DataFrame(data)
```

```python
def vision_language_transformer_demo(
    image_path,
    prompt="a photography of",
    max_new_tokens=30,
    num_beams=5,
    show_table=True
):
    """
    Feature 18: Vision-Language Transformer (BLIP-based)
    
    Demonstrates:
    1. Standard image captioning
    2. Prompt-guided multimodal captioning
    3. Beam-search enhanced captioning
    """
    # Load image
    image = Image.open(image_path).convert("RGB")

    # Display image
    display_image(image, title="Vision-Language Transformer Input")

    # Generate captions
    unconditional_caption = generate_unconditional_caption(
        image=image,
        max_new_tokens=max_new_tokens
    )

    prompt_caption = generate_prompt_guided_caption(
        image=image,
        prompt=prompt,
        max_new_tokens=max_new_tokens
    )

    beam_caption = generate_beam_search_caption(
        image=image,
        max_new_tokens=max_new_tokens,
        num_beams=num_beams
    )

    # Comparison table
    comparison_df = create_caption_comparison_table(
        unconditional_caption,
        prompt_caption,
        beam_caption
    )

    if show_table:
        print("\nCaption Comparison Table:")
        display(comparison_df)

    # Clean output
    print("=" * 90)
    print("VISION-LANGUAGE TRANSFORMER BASED IMAGE CAPTION GENERATION (BLIP)")
    print("=" * 90)

    print("\n[1] Unconditional Caption:")
    print(unconditional_caption)

    print("\n[2] Prompt-Guided Caption:")
    print(prompt_caption)

    print("\n[3] Beam Search Caption:")
    print(beam_caption)

    print("\n[4] Model Interpretation:")
    print("BLIP acts as a practical Vision-Language Transformer by combining visual encoding and language generation.")

    print("\nFeature executed successfully.")

    return {
        "unconditional_caption": unconditional_caption,
        "prompt_guided_caption": prompt_caption,
        "beam_search_caption": beam_caption,
        "comparison_table": comparison_df
    }
```

```python
import warnings
warnings.filterwarnings("ignore")

import re
import torch
import pandas as pd
import matplotlib.pyplot as plt

from PIL import Image
from IPython.display import display
from transformers import BlipProcessor, BlipForConditionalGeneration
from ultralytics import YOLO
```

```python
# Device setup
device = "cuda" if torch.cuda.is_available() else "cpu"

# BLIP model
processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
blip_model = BlipForConditionalGeneration.from_pretrained(
    "Salesforce/blip-image-captioning-base"
).to(device)

# YOLO model
yolo_model = YOLO("yolov8n.pt")

print("BLIP and YOLO loaded successfully.")
print(f"Using device: {device}")
```

```python
def display_image(image, title="Input Image", figsize=(8, 6)):
    plt.figure(figsize=figsize)
    plt.imshow(image)
    plt.axis("off")
    plt.title(title)
    plt.show()
```

```python
def generate_baseline_caption(image, max_new_tokens=30):
    """
    Generate a single baseline caption using greedy decoding.
    """
    inputs = processor(images=image, return_tensors="pt").to(device)

    with torch.no_grad():
        output = blip_model.generate(
            **inputs,
            max_new_tokens=max_new_tokens
        )

    caption = processor.decode(output[0], skip_special_tokens=True).strip()
    return caption
```

```python
def generate_candidate_captions_v2(image, max_new_tokens=30, num_beams=5, num_candidates=5):
    """
    Generate multiple candidate captions using beam search.
    Note: early_stopping removed to avoid warning.
    """
    inputs = processor(images=image, return_tensors="pt").to(device)

    with torch.no_grad():
        outputs = blip_model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            num_beams=max(num_beams, num_candidates),
            num_return_sequences=num_candidates
        )

    captions = []
    for output in outputs:
        caption = processor.decode(output, skip_special_tokens=True).strip()
        captions.append(caption)

    # Remove duplicates while preserving order
    unique_captions = list(dict.fromkeys(captions))
    return unique_captions
```

```python
def detect_objects_v2(image_path, conf_threshold=0.30):
    """
    Detect objects using YOLOv8 and return unique class names.
    """
    results = yolo_model.predict(image_path, conf=conf_threshold, verbose=False)

    detected_objects = []
    if len(results) > 0:
        boxes = results[0].boxes
        names = results[0].names

        if boxes is not None:
            for box in boxes:
                cls_id = int(box.cls[0].item())
                detected_objects.append(names[cls_id])

    detected_objects = list(dict.fromkeys(detected_objects))
    return detected_objects
```

```python
def clean_text_v2(text):
    text = text.lower().strip()
    text = re.sub(r'[^a-z0-9\s]', '', text)
    text = " ".join(text.split())
    return text
```

```python
def compute_caption_reward_v2(caption, detected_objects):
    """
    Improved practical RL-inspired reward function.

    Reward components:
    1. Object match reward
    2. Action word reward
    3. Context keyword reward
    4. Length quality reward
    5. Repetition penalty
    6. Generic caption penalty
    """
    caption_clean = clean_text_v2(caption)
    words = caption_clean.split()
    word_set = set(words)

    # -----------------------------
    # 1) Object match reward
    # -----------------------------
    matched_objects = []
    for obj in detected_objects:
        obj_clean = obj.lower()
        if obj_clean in word_set:
            matched_objects.append(obj_clean)

    object_reward = len(set(matched_objects)) * 2.0

    # -----------------------------
    # 2) Action word reward (stronger than v1)
    # -----------------------------
    action_words = {
        "playing", "riding", "running", "walking", "standing",
        "sitting", "holding", "jumping", "driving", "eating",
        "drinking", "looking", "posing", "swimming", "flying"
    }

    matched_actions = word_set.intersection(action_words)
    action_reward = min(len(matched_actions) * 1.5, 3.0)

    # -----------------------------
    # 3) Context keyword reward (important for better semantic richness)
    # -----------------------------
    context_keywords = {
        "accordion", "guitar", "piano", "violin", "drum", "instrument",
        "ball", "bench", "bicycle", "motorcycle", "car", "truck",
        "dog", "cat", "horse", "water", "street", "field", "fountain",
        "table", "bottle", "toy", "camera"
    }

    matched_context = word_set.intersection(context_keywords)
    context_reward = min(len(matched_context) * 1.2, 3.0)

    # -----------------------------
    # 4) Length quality reward
    # Ideal range: 4 to 14 words
    # -----------------------------
    word_count = len(words)
    if 4 <= word_count <= 14:
        length_reward = 2.0
    elif 3 <= word_count < 4:
        length_reward = 1.0
    elif word_count > 14:
        length_reward = 0.5
    else:
        length_reward = 0.0

    # -----------------------------
    # 5) Repetition penalty
    # -----------------------------
    unique_word_ratio = len(set(words)) / max(len(words), 1)
    repetition_penalty = 0.0
    if unique_word_ratio < 0.70:
        repetition_penalty = -1.5

    # -----------------------------
    # 6) Generic caption penalty
    # Penalize overly generic captions if richer captions exist
    # -----------------------------
    generic_phrases = {
        "a person",
        "a man",
        "a woman",
        "a group of people",
        "a crowd of people",
        "a dog",
        "a cat"
    }

    generic_penalty = -1.0 if caption_clean in generic_phrases else 0.0

    # -----------------------------
    # Final total reward
    # -----------------------------
    total_reward = (
        object_reward
        + action_reward
        + context_reward
        + length_reward
        + repetition_penalty
        + generic_penalty
    )

    return {
        "caption": caption,
        "object_reward": round(object_reward, 2),
        "action_reward": round(action_reward, 2),
        "context_reward": round(context_reward, 2),
        "length_reward": round(length_reward, 2),
        "repetition_penalty": round(repetition_penalty, 2),
        "generic_penalty": round(generic_penalty, 2),
        "total_reward": round(total_reward, 2)
    }
```

```python
def score_candidate_captions_v2(candidates, detected_objects):
    """
    Score all candidate captions with detailed reward breakdown.
    """
    scored_results = []

    for caption in candidates:
        reward_info = compute_caption_reward_v2(caption, detected_objects)
        scored_results.append(reward_info)

    # Sort by total reward descending
    scored_results = sorted(scored_results, key=lambda x: x["total_reward"], reverse=True)
    return scored_results
```

```python
def select_best_caption_v2(scored_results):
    """
    Select best caption using:
    1. Highest total reward
    2. If tie -> prefer longer (more informative) caption
    """
    if not scored_results:
        return ""

    max_reward = max(item["total_reward"] for item in scored_results)

    top_candidates = [item for item in scored_results if item["total_reward"] == max_reward]

    # Tie-break by caption length (prefer richer caption)
    best_item = max(top_candidates, key=lambda x: len(x["caption"].split()))
    return best_item["caption"]
```

```python
def create_reward_breakdown_table_v2(scored_results):
    """
    Create a detailed reward breakdown table.
    """
    df = pd.DataFrame(scored_results)
    return df[[
        "caption",
        "object_reward",
        "action_reward",
        "context_reward",
        "length_reward",
        "repetition_penalty",
        "generic_penalty",
        "total_reward"
    ]]
```

```python
def reinforcement_learning_optimization_v2(
    image_path,
    max_new_tokens=30,
    num_beams=6,
    num_candidates=5,
    conf_threshold=0.30,
    show_table=True
):
    """
    FINAL v2:
    Reinforcement Learning Inspired Caption Optimization
    (Reward-Based Refinement with Better Reward Design)
    """
    # Load image
    image = Image.open(image_path).convert("RGB")

    # Display image
    display_image(image, title="Reinforcement Learning Optimization Input")

    # Step 1: Baseline caption
    baseline_caption = generate_baseline_caption(
        image=image,
        max_new_tokens=max_new_tokens
    )

    # Step 2: Candidate captions
    candidate_captions = generate_candidate_captions_v2(
        image=image,
        max_new_tokens=max_new_tokens,
        num_beams=num_beams,
        num_candidates=num_candidates
    )

    # Ensure baseline is included in candidate set
    if baseline_caption not in candidate_captions:
        candidate_captions.insert(0, baseline_caption)

    # Remove duplicates again after insertion
    candidate_captions = list(dict.fromkeys(candidate_captions))

    # Step 3: Detect objects
    detected_objects = detect_objects_v2(
        image_path=image_path,
        conf_threshold=conf_threshold
    )

    # Step 4: Score candidates
    scored_results = score_candidate_captions_v2(
        candidates=candidate_captions,
        detected_objects=detected_objects
    )

    # Step 5: Select best optimized caption
    optimized_caption = select_best_caption_v2(scored_results)

    # Step 6: Improvement score
    baseline_score = None
    for item in scored_results:
        if item["caption"] == baseline_caption:
            baseline_score = item["total_reward"]
            break

    optimized_score = scored_results[0]["total_reward"] if scored_results else 0.0
    improvement = round(optimized_score - (baseline_score if baseline_score is not None else 0.0), 2)

    # Table
    reward_df = create_reward_breakdown_table_v2(scored_results)

    if show_table:
        print("\nReward Breakdown Table:")
        display(reward_df)

    # Clean output
    print("=" * 100)
    print("REINFORCEMENT LEARNING INSPIRED CAPTION OPTIMIZATION (FINAL v2)")
    print("=" * 100)

    print("\n[1] Baseline Caption:")
    print(baseline_caption)

    print("\n[2] Detected Objects (YOLO):")
    if detected_objects:
        print(", ".join(detected_objects))
    else:
        print("No major objects detected")

    print("\n[3] Candidate Captions:")
    for i, cap in enumerate(candidate_captions, 1):
        print(f"{i}. {cap}")

    print("\n[4] Optimized Caption:")
    print(optimized_caption)

    print("\n[5] Reward Improvement:")
    print(f"Baseline Reward: {baseline_score}")
    print(f"Optimized Reward: {optimized_score}")
    print(f"Improvement: {improvement}")

    print("\n[6] Model Interpretation:")
    print("A reinforcement-learning-inspired multi-factor reward function refines caption selection by preferring object-consistent, action-aware, and semantically richer captions.")

    print("\nFeature executed successfully.")

    return {
        "baseline_caption": baseline_caption,
        "detected_objects": detected_objects,
        "candidate_captions": candidate_captions,
        "optimized_caption": optimized_caption,
        "reward_table": reward_df,
        "baseline_reward": baseline_score,
        "optimized_reward": optimized_score,
        "improvement": improvement
    }
```

```python
# Device setup
device = "cuda" if torch.cuda.is_available() else "cpu"

# BLIP model
processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
blip_model = BlipForConditionalGeneration.from_pretrained(
    "Salesforce/blip-image-captioning-base"
).to(device)

# YOLO model
yolo_model = YOLO("yolov8n.pt")

print("BLIP and YOLO loaded successfully.")
print(f"Using device: {device}")
```

```python
def display_image(image, title="Input Image", figsize=(8, 6)):
    plt.figure(figsize=figsize)
    plt.imshow(image)
    plt.axis("off")
    plt.title(title)
    plt.show()
```

```python
def generate_base_caption(image, max_new_tokens=30):
    """
    Generate a standard base caption using BLIP.
    """
    inputs = processor(images=image, return_tensors="pt").to(device)

    with torch.no_grad():
        output = blip_model.generate(
            **inputs,
            max_new_tokens=max_new_tokens
        )

    caption = processor.decode(output[0], skip_special_tokens=True).strip()
    return caption
```

```python
def detect_objects_for_domain(image_path, conf_threshold=0.30):
    """
    Detect major objects using YOLO and return unique class names.
    """
    results = yolo_model.predict(image_path, conf=conf_threshold, verbose=False)

    detected_objects = []
    if len(results) > 0:
        boxes = results[0].boxes
        names = results[0].names

        if boxes is not None:
            for box in boxes:
                cls_id = int(box.cls[0].item())
                detected_objects.append(names[cls_id])

    detected_objects = list(dict.fromkeys(detected_objects))
    return detected_objects
```

```python
def build_object_summary(detected_objects):
    """
    Create a short object summary string.
    """
    if not detected_objects:
        return "no major objects detected"

    if len(detected_objects) == 1:
        return detected_objects[0]

    if len(detected_objects) <= 4:
        return ", ".join(detected_objects)

    return ", ".join(detected_objects[:4]) + ", etc."
```

```python
def detect_basic_hazard_context(base_caption, detected_objects):
    """
    Simple practical hazard inference based on caption + detected objects.
    """
    caption_lower = base_caption.lower()
    objects_lower = [obj.lower() for obj in detected_objects]

    hazard_keywords = {
        "fire", "smoke", "knife", "accident", "crash", "spill", "flood",
        "helmet", "traffic", "road", "street", "water", "motorcycle", "car", "truck"
    }

    found = []

    for word in hazard_keywords:
        if word in caption_lower or word in objects_lower:
            found.append(word)

    return list(dict.fromkeys(found))
```

```python
def generate_domain_adaptive_caption(base_caption, detected_objects, mode="general"):
    """
    Adapt caption according to selected domain mode.
    
    Supported modes:
    - general
    - surveillance
    - assistive
    - social_media
    - ecommerce
    - safety
    """
    mode = mode.lower().strip()
    object_summary = build_object_summary(detected_objects)
    hazard_context = detect_basic_hazard_context(base_caption, detected_objects)

    # -----------------------------
    # General Mode
    # -----------------------------
    if mode == "general":
        final_caption = base_caption

    # -----------------------------
    # Surveillance Mode
    # -----------------------------
    elif mode == "surveillance":
        final_caption = (
            f"Surveillance observation: {base_caption}. "
            f"Detected objects include {object_summary}."
        )

    # -----------------------------
    # Assistive Mode
    # -----------------------------
    elif mode == "assistive":
        final_caption = (
            f"Assistive description: {base_caption}. "
            f"The main visible objects are {object_summary}."
        )

    # -----------------------------
    # Social Media Mode
    # -----------------------------
    elif mode == "social_media":
        final_caption = (
            f"{base_caption.capitalize()} — a visually engaging moment captured in the scene."
        )

    # -----------------------------
    # E-commerce Mode
    # -----------------------------
    elif mode == "ecommerce":
        if detected_objects:
            final_caption = (
                f"Product-focused description: {base_caption}. "
                f"Prominent visible items include {object_summary}."
            )
        else:
            final_caption = (
                f"Product-focused description: {base_caption}."
            )

    # -----------------------------
    # Safety / Hazard Mode
    # -----------------------------
    elif mode == "safety":
        if hazard_context:
            final_caption = (
                f"Safety-aware description: {base_caption}. "
                f"Potential caution-related context detected: {', '.join(hazard_context)}."
            )
        else:
            final_caption = (
                f"Safety-aware description: {base_caption}. "
                f"No obvious hazard indicators detected in the scene."
            )

    # -----------------------------
    # Fallback
    # -----------------------------
    else:
        final_caption = base_caption

    return final_caption
```

```python
def generate_all_domain_modes(base_caption, detected_objects):
    """
    Generate captions for all supported domain modes.
    """
    modes = [
        "general",
        "surveillance",
        "assistive",
        "social_media",
        "ecommerce",
        "safety"
    ]

    results = []

    for mode in modes:
        adapted_caption = generate_domain_adaptive_caption(
            base_caption=base_caption,
            detected_objects=detected_objects,
            mode=mode
        )

        results.append({
            "Domain Mode": mode,
            "Adapted Caption": adapted_caption
        })

    return pd.DataFrame(results)
```

```python
def domain_adaptive_caption_modes_demo(
    image_path,
    selected_mode="general",
    max_new_tokens=30,
    conf_threshold=0.30,
    show_all_modes=True
):
    """
    Feature 27:
    Domain-Adaptive Caption Modes (Context-Specific Caption Generation)
    """
    # Load image
    image = Image.open(image_path).convert("RGB")

    # Display image
    display_image(image, title="Domain-Adaptive Caption Modes Input")

    # Step 1: Base caption
    base_caption = generate_base_caption(
        image=image,
        max_new_tokens=max_new_tokens
    )

    # Step 2: Detect objects
    detected_objects = detect_objects_for_domain(
        image_path=image_path,
        conf_threshold=conf_threshold
    )

    # Step 3: Generate selected domain caption
    domain_caption = generate_domain_adaptive_caption(
        base_caption=base_caption,
        detected_objects=detected_objects,
        mode=selected_mode
    )

    # Step 4: Optional comparison table for all modes
    comparison_df = None
    if show_all_modes:
        comparison_df = generate_all_domain_modes(
            base_caption=base_caption,
            detected_objects=detected_objects
        )
        print("\nDomain Mode Comparison Table:")
        display(comparison_df)

    # Clean output
    print("=" * 100)
    print("DOMAIN-ADAPTIVE CAPTION MODES (CONTEXT-SPECIFIC CAPTION GENERATION)")
    print("=" * 100)

    print("\n[1] Base Caption:")
    print(base_caption)

    print("\n[2] Detected Objects (YOLO):")
    if detected_objects:
        print(", ".join(detected_objects))
    else:
        print("No major objects detected")

    print(f"\n[3] Selected Domain Mode: {selected_mode}")
    print(domain_caption)

    print("\n[4] Model Interpretation:")
    print("The same visual content is reformulated differently based on the selected application domain.")

    print("\nFeature executed successfully.")

    return {
        "base_caption": base_caption,
        "detected_objects": detected_objects,
        "selected_mode": selected_mode,
        "domain_adaptive_caption": domain_caption,
        "comparison_table": comparison_df
    }
```

```python
def clean_caption_text(text):
    """
    Clean and normalize caption text.
    """
    text = text.strip().lower()
    text = re.sub(r"\s+", " ", text)
    return text


def generate_blip_caption(image):
    """
    Generate caption using BLIP with beam search.
    """
    inputs = processor(images=image, return_tensors="pt").to(device)

    output_ids = blip_model.generate(
        **inputs,
        max_new_tokens=30,
        num_beams=5,
        early_stopping=True
    )

    caption = processor.decode(output_ids[0], skip_special_tokens=True)
    return caption


def generate_blip_caption_with_prompt(image, prompt="a photo of"):
    """
    Prompted caption generation (can improve results in some scenes).
    """
    inputs = processor(images=image, text=prompt, return_tensors="pt").to(device)

    output_ids = blip_model.generate(
        **inputs,
        max_new_tokens=30,
        num_beams=5,
        early_stopping=True
    )

    caption = processor.decode(output_ids[0], skip_special_tokens=True)
    return caption
```

```python
def analyze_hazards(caption):
    """
    Analyze caption for hazards and produce:
    - detected hazards
    - severity score (0-100)
    - risk level
    - warnings
    - recommendations
    - confidence estimate
    """
    caption = clean_caption_text(caption)

    detected_hazards = []
    warnings = []
    recommendations = []
    severity_scores = []

    # Match hazards
    for hazard_name, rule in HAZARD_RULES.items():
        matched = False
        for keyword in rule["keywords"]:
            if keyword in caption:
                matched = True
                break

        if matched:
            detected_hazards.append(hazard_name)
            warnings.append(rule["warning"])
            recommendations.append(rule["recommendation"])
            severity_scores.append(rule["severity"])

    # No hazards found
    if not detected_hazards:
        return {
            "detected_hazards": [],
            "severity_score": 10,
            "risk_level": "LOW",
            "warnings": ["No obvious hazards detected from the scene."],
            "recommendations": ["Maintain normal situational awareness."],
            "confidence": 0.55
        }

    # Aggregate severity
    # Weighted average + slight boost for multiple hazards
    base_score = sum(severity_scores) / len(severity_scores)
    multi_hazard_bonus = min((len(detected_hazards) - 1) * 7, 20)
    final_score = min(int(base_score + multi_hazard_bonus), 100)

    # Map to risk level
    if final_score >= 85:
        risk_level = "CRITICAL"
    elif final_score >= 70:
        risk_level = "HIGH"
    elif final_score >= 40:
        risk_level = "MEDIUM"
    else:
        risk_level = "LOW"

    # Confidence estimate (simple rule-based heuristic)
    confidence = min(0.60 + (0.08 * len(detected_hazards)), 0.95)

    return {
        "detected_hazards": detected_hazards,
        "severity_score": final_score,
        "risk_level": risk_level,
        "warnings": warnings,
        "recommendations": recommendations,
        "confidence": round(confidence, 2)
    }
```

```python
def create_hazard_aware_caption(base_caption, hazard_result):
    """
    Build final advanced hazard-aware caption.
    """
    detected = hazard_result["detected_hazards"]
    risk_level = hazard_result["risk_level"]
    severity_score = hazard_result["severity_score"]
    warnings = hazard_result["warnings"]
    recommendations = hazard_result["recommendations"]
    confidence = hazard_result["confidence"]

    if detected:
        hazard_text = ", ".join([h.replace("_", " ") for h in detected])
        warning_text = warnings[0] if warnings else "Potential hazards detected."
        recommendation_text = recommendations[0] if recommendations else "Maintain caution."

        final_text = (
            f"Scene description: {base_caption}. "
            f"Potential hazards detected: {hazard_text}. "
            f"Risk level: {risk_level} (Severity Score: {severity_score}/100). "
            f"Warning: {warning_text} "
            f"Recommended action: {recommendation_text} "
            f"System confidence: {confidence:.2f}."
        )
    else:
        final_text = (
            f"Scene description: {base_caption}. "
            f"No major hazards identified. "
            f"Risk level: {risk_level} (Severity Score: {severity_score}/100). "
            f"Recommended action: Maintain normal situational awareness. "
            f"System confidence: {confidence:.2f}."
        )

    return final_text
```

```python
def log_result_to_csv(file_name, caption, hazard_result, final_caption):
    """
    Save result to CSV log.
    """
    row = {
        "timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        "file_name": file_name,
        "generated_caption": caption,
        "detected_hazards": ", ".join(hazard_result["detected_hazards"]) if hazard_result["detected_hazards"] else "None",
        "severity_score": hazard_result["severity_score"],
        "risk_level": hazard_result["risk_level"],
        "warnings": " | ".join(hazard_result["warnings"]),
        "recommendations": " | ".join(hazard_result["recommendations"]),
        "confidence": hazard_result["confidence"],
        "final_hazard_caption": final_caption
    }

    df_row = pd.DataFrame([row])

    if os.path.exists(CSV_LOG_PATH):
        df_row.to_csv(CSV_LOG_PATH, mode='a', header=False, index=False)
    else:
        df_row.to_csv(CSV_LOG_PATH, mode='w', header=True, index=False)

    return row
```

```python
def get_risk_color(risk_level):
    color_map = {
        "LOW": "#2e7d32",       # green
        "MEDIUM": "#f9a825",    # yellow/orange
        "HIGH": "#ef6c00",      # orange-red
        "CRITICAL": "#c62828"   # red
    }
    return color_map.get(risk_level, "#424242")
```

```python
upload = widgets.FileUpload(
    accept='image/*',
    multiple=False,
    description='Upload Image'
)

output = widgets.Output()

def on_upload_change(change):
    with output:
        output.clear_output()

        if upload.value:
            # Compatibility for different ipywidgets versions
            if isinstance(upload.value, tuple):
                uploaded_file = upload.value[0]
                file_name = uploaded_file['name']
                content = uploaded_file['content']
            else:
                uploaded_file = list(upload.value.values())[0]
                file_name = uploaded_file['metadata']['name']
                content = uploaded_file['content']

            # Load image
            image = Image.open(io.BytesIO(content)).convert("RGB")

            # Save uploaded image
            save_path = os.path.join(UPLOAD_FOLDER, file_name)
            with open(save_path, "wb") as f:
                f.write(content)

            # Show image
            display(HTML(f"""
            <h2 style='color:#1565c0;'>📷 Uploaded Image</h2>
            <p><b>File Name:</b> {file_name}</p>
            <p><b>Saved To:</b> {save_path}</p>
            """))

            plt.figure(figsize=(8, 6))
            plt.imshow(image)
            plt.axis("off")
            plt.title("Uploaded Image")
            plt.show()

            try:
                # -----------------------------
                # Generate captions (2 ways)
                # -----------------------------
                caption1 = generate_blip_caption(image)
                caption2 = generate_blip_caption_with_prompt(image, prompt="a photo of")

                # Pick the longer / more informative one
                base_caption = caption2 if len(caption2.split()) > len(caption1.split()) else caption1
                base_caption = base_caption.strip()

                # -----------------------------
                # Hazard analysis
                # -----------------------------
                hazard_result = analyze_hazards(base_caption)

                # -----------------------------
                # Final advanced hazard caption
                # -----------------------------
                final_caption = create_hazard_aware_caption(base_caption, hazard_result)

                # -----------------------------
                # Log to CSV
                # -----------------------------
                log_result_to_csv(file_name, base_caption, hazard_result, final_caption)

                # -----------------------------
                # Display results
                # -----------------------------
                risk_level = hazard_result["risk_level"]
                severity_score = hazard_result["severity_score"]
                confidence = hazard_result["confidence"]
                risk_color = get_risk_color(risk_level)

                display(HTML(f"""
                <h2 style='color:#2e7d32;'>📝 Generated Caption</h2>
                <div style='padding:10px; border-radius:8px; background:#f1f8e9; font-size:16px;'>
                    {base_caption}
                </div>
                """))

                display(HTML(f"""
                <h2 style='color:#ef6c00;'>⚠️ Hazard Analysis</h2>
                <div style='padding:12px; border-radius:8px; background:#fff3e0; font-size:15px;'>
                    <p><b>Detected Hazards:</b> {", ".join(hazard_result["detected_hazards"]) if hazard_result["detected_hazards"] else "None"}</p>
                    <p><b>Severity Score:</b> {severity_score}/100</p>
                    <p><b>Risk Level:</b> <span style='color:{risk_color}; font-weight:bold;'>{risk_level}</span></p>
                    <p><b>System Confidence:</b> {confidence:.2f}</p>
                </div>
                """))

                display(HTML("<h3>🚨 Warnings</h3>"))
                for w in hazard_result["warnings"]:
                    print("-", w)

                display(HTML("<h3>✅ Recommendations</h3>"))
                for r in hazard_result["recommendations"]:
                    print("-", r)

                display(HTML(f"""
                <h2 style='color:#c62828;'>🚨 Final Risk / Hazard Awareness Caption</h2>
                <div style='padding:12px; border-radius:8px; background:#ffebee; font-size:16px;'>
                    {final_caption}
                </div>
                """))

                display(HTML(f"""
                <p style='color:#616161;'><b>CSV Log Updated:</b> {CSV_LOG_PATH}</p>
                """))

            except Exception as e:
                display(HTML(f"""
                <h3 style='color:red;'>❌ Error during processing</h3>
                <p>{str(e)}</p>
                """))

upload.observe(on_upload_change, names='value')

display(HTML("<h1 style='color:#0d47a1;'>📤 Risk / Hazard Awareness Captioning System</h1>"))
display(HTML("<p>Upload an image to generate a scene caption, detect hazards, estimate severity, and receive safety recommendations.</p>"))
display(upload)
display(output)
```

```python
def clean_text(text):
    text = text.strip().lower()
    text = re.sub(r"\s+", " ", text)
    return text

def keyword_match(text, keyword):
    pattern = r'\b' + re.escape(keyword.lower()) + r'\b'
    return re.search(pattern, text.lower()) is not None

def unique_preserve_order(items):
    seen = set()
    result = []
    for item in items:
        if item not in seen:
            seen.add(item)
            result.append(item)
    return result
```

```python
def generate_blip_caption(image):
    inputs = processor(images=image, return_tensors="pt").to(device)

    output_ids = blip_model.generate(
        **inputs,
        max_new_tokens=30,
        num_beams=5,
        early_stopping=True
    )

    caption = processor.decode(output_ids[0], skip_special_tokens=True)
    return caption


def generate_blip_caption_with_prompt(image, prompt="a photo of"):
    inputs = processor(images=image, text=prompt, return_tensors="pt").to(device)

    output_ids = blip_model.generate(
        **inputs,
        max_new_tokens=30,
        num_beams=5,
        early_stopping=True
    )

    caption = processor.decode(output_ids[0], skip_special_tokens=True)
    return caption


def generate_best_caption(image):
    prompts = [
        None,
        "a photo of an outdoor scene",
        "a photo of an indoor scene",
        "a photo of people in a public place",
        "a photo of a street scene",
        "a photo of a park or playground",
        "a photo of a sports area",
        "a photo of a room or office",
        "a photo of a beach or water scene"
    ]

    captions = []

    for prompt in prompts:
        try:
            if prompt is None:
                cap = generate_blip_caption(image)
            else:
                cap = generate_blip_caption_with_prompt(image, prompt=prompt)

            if cap and cap not in captions:
                captions.append(cap)
        except:
            pass

    preferred_keywords = [
        "street", "road", "park", "playground", "beach", "ocean", "water", "pool",
        "office", "classroom", "kitchen", "bedroom", "sofa", "stadium", "football",
        "train", "airport", "camera", "tripod", "market", "store", "restaurant"
    ]

    best_caption = captions[0] if captions else "a scene"
    best_score = -1

    for cap in captions:
        score = len(cap.split())
        for kw in preferred_keywords:
            if kw in cap.lower():
                score += 3
        if "people" in cap.lower():
            score += 2
        if "outdoor" in cap.lower() or "indoor" in cap.lower():
            score += 2

        if score > best_score:
            best_score = score
            best_caption = cap

    return best_caption
```

```python
def infer_location(caption):
    caption = clean_text(caption)

    matched_locations = []
    matched_categories = []
    descriptions = []
    confidences = []

    for location_name, rule in LOCATION_RULES.items():
        matched = False
        for keyword in rule["keywords"]:
            if keyword_match(caption, keyword):
                matched = True
                break

        if matched:
            matched_locations.append(location_name)
            matched_categories.append(rule["category"])
            descriptions.append(rule["description"])
            confidences.append(rule["confidence"])

    matched_locations = unique_preserve_order(matched_locations)
    matched_categories = unique_preserve_order(matched_categories)
    descriptions = unique_preserve_order(descriptions)

    # Fallback if nothing matched
    if not matched_locations:
        # heuristic indoor/outdoor fallback
        outdoor_terms = ["tree", "road", "car", "street", "grass", "sky", "park", "beach", "water"]
        indoor_terms = ["room", "chair", "table", "bed", "sofa", "desk", "kitchen", "computer"]

        outdoor_score = sum([1 for term in outdoor_terms if keyword_match(caption, term)])
        indoor_score = sum([1 for term in indoor_terms if keyword_match(caption, term)])

        if outdoor_score > indoor_score:
            estimated_place = "general_outdoor_area"
            category = "Outdoor / General"
            description = "Likely an outdoor environment, but exact location is uncertain."
            confidence = 0.55
        elif indoor_score > outdoor_score:
            estimated_place = "general_indoor_area"
            category = "Indoor / General"
            description = "Likely an indoor environment, but exact location is uncertain."
            confidence = 0.55
        else:
            estimated_place = "unknown_scene"
            category = "Unknown / General"
            description = "Unable to confidently infer a specific location from the visual caption."
            confidence = 0.45

        return {
            "estimated_place": estimated_place,
            "location_category": category,
            "matched_locations": [],
            "description": description,
            "confidence": round(confidence, 2)
        }

    # choose best location
    best_idx = confidences.index(max(confidences))
    estimated_place = matched_locations[best_idx]
    best_category = matched_categories[best_idx]
    best_description = descriptions[best_idx]

    # boost if multiple matches
    final_confidence = min(max(confidences) + (0.03 * (len(matched_locations) - 1)), 0.97)

    return {
        "estimated_place": estimated_place,
        "location_category": best_category,
        "matched_locations": matched_locations,
        "description": best_description,
        "confidence": round(final_confidence, 2)
    }
```

```python
def create_location_inference_caption(base_caption, location_result):
    estimated_place = location_result["estimated_place"].replace("_", " ").title()
    location_category = location_result["location_category"]
    matched_locations = location_result["matched_locations"]
    description = location_result["description"]
    confidence = location_result["confidence"]

    if matched_locations:
        matched_text = ", ".join([loc.replace("_", " ") for loc in matched_locations])
    else:
        matched_text = "No strong rule-based scene matches"

    final_text = (
        f"Scene description: {base_caption}. "
        f"Estimated location: {estimated_place}. "
        f"Location category: {location_category}. "
        f"Matched scene cues: {matched_text}. "
        f"Inference summary: {description} "
        f"System confidence: {confidence:.2f}."
    )

    return final_text
```

```python
def get_location_color(category):
    category = category.lower()

    if "outdoor" in category and "natural" in category:
        return "#2e7d32"
    elif "urban" in category or "transport" in category:
        return "#1565c0"
    elif "indoor" in category:
        return "#6a1b9a"
    elif "commercial" in category:
        return "#ef6c00"
    elif "sports" in category:
        return "#00897b"
    else:
        return "#455a64"


def render_location_ui(image_path, base_caption, location_result, final_text):
    estimated_place = location_result["estimated_place"].replace("_", " ").title()
    location_category = location_result["location_category"]
    matched_locations = location_result["matched_locations"]
    description = location_result["description"]
    confidence = location_result["confidence"]

    color = get_location_color(location_category)

    image = Image.open(image_path).convert("RGB")
    plt.figure(figsize=(8, 6))
    plt.imshow(image)
    plt.axis("off")
    plt.title("Uploaded Image")
    plt.show()

    matched_text = ", ".join([loc.replace("_", " ") for loc in matched_locations]) if matched_locations else "No strong matches"

    html = f"""
    <div style="font-family: Arial, sans-serif; padding: 12px;">

        <h2 style="color:#2e7d32;">📝 Generated Caption</h2>
        <div style="background:#f1f8e9; padding:12px; border-radius:10px; font-size:16px;">
            {base_caption}
        </div>

        <h2 style="color:{color}; margin-top:20px;">📍 Location Inference Result</h2>
        <div style="background:#f5f5f5; padding:14px; border-radius:12px; border-left:6px solid {color};">
            <p><b>Estimated Place:</b> {estimated_place}</p>
            <p><b>Location Category:</b> <span style="color:{color}; font-weight:bold;">{location_category}</span></p>
            <p><b>Matched Scene Cues:</b> {matched_text}</p>
            <p><b>Inference Summary:</b> {description}</p>
            <p><b>System Confidence:</b> {confidence}</p>
        </div>

        <h2 style="color:#c62828; margin-top:20px;">🧠 Final Visual Location Inference Caption</h2>
        <div style="background:#e3f2fd; padding:14px; border-radius:12px; font-size:16px;">
            {final_text}
        </div>
    </div>
    """

    display(HTML(html))
```

```python
def run_visual_location_inference(image_path):
    if not os.path.exists(image_path):
        print("❌ Image path not found:", image_path)
        return

    print("🧠 Generating caption...")
    image = Image.open(image_path).convert("RGB")

    base_caption = generate_best_caption(image)
    print("✅ Caption generated")

    print("📍 Inferring location...")
    location_result = infer_location(base_caption)

    final_text = create_location_inference_caption(base_caption, location_result)

    render_location_ui(image_path, base_caption, location_result, final_text)
```

```python
def dms_to_decimal(dms, ref):
    try:
        degrees = float(dms[0][0]) / float(dms[0][1])
        minutes = float(dms[1][0]) / float(dms[1][1])
        seconds = float(dms[2][0]) / float(dms[2][1])
        dec = degrees + minutes / 60 + seconds / 3600
        if ref in ['S', 'W']:
            dec = -dec
        return dec
    except:
        return None

def extract_exif_gps(image_path):
    try:
        img = Image.open(image_path)
        exif_data = img._getexif()
        if not exif_data:
            return None, None

        exif = {}
        for tag_id, value in exif_data.items():
            tag = ExifTags.TAGS.get(tag_id, tag_id)
            exif[tag] = value

        gps_info = exif.get("GPSInfo", None)
        if not gps_info:
            return None, None

        gps_parsed = {}
        for key in gps_info.keys():
            decoded = ExifTags.GPSTAGS.get(key, key)
            gps_parsed[decoded] = gps_info[key]

        if all(k in gps_parsed for k in ["GPSLatitude", "GPSLatitudeRef", "GPSLongitude", "GPSLongitudeRef"]):
            lat = dms_to_decimal(gps_parsed["GPSLatitude"], gps_parsed["GPSLatitudeRef"])
            lon = dms_to_decimal(gps_parsed["GPSLongitude"], gps_parsed["GPSLongitudeRef"])
            return lat, lon

        return None, None
    except:
        return None, None

def reverse_geocode(lat, lon):
    try:
        url = "https://nominatim.openstreetmap.org/reverse"
        params = {
            "format": "jsonv2",
            "lat": lat,
            "lon": lon,
            "zoom": 18,
            "addressdetails": 1
        }
        headers = {"User-Agent": "GeoCaptionFinalPro/1.0"}
        r = requests.get(url, params=params, headers=headers, timeout=10)
        r.raise_for_status()
        data = r.json()
        return data.get("display_name", "Unknown location")
    except:
        return None

def generate_caption(image):
    try:
        inputs = blip_processor(images=image, return_tensors="pt").to(DEVICE)
        out = blip_model.generate(**inputs, max_new_tokens=40)
        caption = blip_processor.decode(out[0], skip_special_tokens=True)
        return caption
    except:
        return None

def clip_rank_prompts(image, prompts):
    try:
        image_tensor = clip_preprocess(image).unsqueeze(0).to(DEVICE)
        text_tokens = clip_tokenizer(prompts).to(DEVICE)

        with torch.no_grad():
            image_features = clip_model.encode_image(image_tensor)
            text_features = clip_model.encode_text(text_tokens)

            image_features /= image_features.norm(dim=-1, keepdim=True)
            text_features /= text_features.norm(dim=-1, keepdim=True)

            probs = (100.0 * image_features @ text_features.T).softmax(dim=-1)[0].cpu().numpy()

        ranked = sorted(zip(prompts, probs), key=lambda x: x[1], reverse=True)
        return ranked
    except:
        return []

def run_yolo(image_path):
    if yolo_model is None:
        return [], None

    try:
        results = yolo_model(image_path, verbose=False)
        r = results[0]

        detections = []
        if r.boxes is not None and len(r.boxes) > 0:
            for box in r.boxes:
                cls_id = int(box.cls[0].item())
                conf = float(box.conf[0].item())
                label = r.names.get(cls_id, str(cls_id))
                detections.append((label, conf))

        annotated = r.plot()
        annotated = annotated[:, :, ::-1]  # BGR -> RGB
        return detections, annotated
    except:
        return [], None

def suspicious_manual_gps(lat, lon):
    if lat is None or lon is None:
        return False

    # exact integer coordinates often indicate dummy/test
    if abs(lat - round(lat)) < 1e-9 and abs(lon - round(lon)) < 1e-9:
        return True

    # invalid bounds
    if not (-90 <= lat <= 90 and -180 <= lon <= 180):
        return True

    return False

def should_show_map(lat, lon, gps_source):
    if lat is None or lon is None:
        return False
    if gps_source == "manual_override" and suspicious_manual_gps(lat, lon):
        return False
    return True
```

```python
def dms_to_decimal(dms, ref):
    try:
        degrees = float(dms[0][0]) / float(dms[0][1])
        minutes = float(dms[1][0]) / float(dms[1][1])
        seconds = float(dms[2][0]) / float(dms[2][1])
        dec = degrees + minutes / 60 + seconds / 3600
        if ref in ['S', 'W']:
            dec = -dec
        return dec
    except:
        return None

def extract_exif_gps(image_path):
    try:
        img = Image.open(image_path)
        exif_data = img._getexif()
        if not exif_data:
            return None, None

        exif = {}
        for tag_id, value in exif_data.items():
            tag = ExifTags.TAGS.get(tag_id, tag_id)
            exif[tag] = value

        gps_info = exif.get("GPSInfo", None)
        if not gps_info:
            return None, None

        gps_parsed = {}
        for key in gps_info.keys():
            decoded = ExifTags.GPSTAGS.get(key, key)
            gps_parsed[decoded] = gps_info[key]

        if all(k in gps_parsed for k in ["GPSLatitude", "GPSLatitudeRef", "GPSLongitude", "GPSLongitudeRef"]):
            lat = dms_to_decimal(gps_parsed["GPSLatitude"], gps_parsed["GPSLatitudeRef"])
            lon = dms_to_decimal(gps_parsed["GPSLongitude"], gps_parsed["GPSLongitudeRef"])
            return lat, lon

        return None, None
    except:
        return None, None

def reverse_geocode(lat, lon):
    try:
        url = "https://nominatim.openstreetmap.org/reverse"
        params = {
            "format": "jsonv2",
            "lat": lat,
            "lon": lon,
            "zoom": 18,
            "addressdetails": 1
        }
        headers = {"User-Agent": "GeoCaptionFinalPro/1.0"}
        r = requests.get(url, params=params, headers=headers, timeout=10)
        r.raise_for_status()
        data = r.json()
        return data.get("display_name", "Unknown location")
    except:
        return None

def generate_caption(image):
    try:
        inputs = blip_processor(images=image, return_tensors="pt").to(DEVICE)
        out = blip_model.generate(**inputs, max_new_tokens=40)
        caption = blip_processor.decode(out[0], skip_special_tokens=True)
        return caption
    except:
        return None

def clip_rank_prompts(image, prompts):
    try:
        image_tensor = clip_preprocess(image).unsqueeze(0).to(DEVICE)
        text_tokens = clip_tokenizer(prompts).to(DEVICE)

        with torch.no_grad():
            image_features = clip_model.encode_image(image_tensor)
            text_features = clip_model.encode_text(text_tokens)

            image_features /= image_features.norm(dim=-1, keepdim=True)
            text_features /= text_features.norm(dim=-1, keepdim=True)

            probs = (100.0 * image_features @ text_features.T).softmax(dim=-1)[0].cpu().numpy()

        ranked = sorted(zip(prompts, probs), key=lambda x: x[1], reverse=True)
        return ranked
    except:
        return []

def run_yolo(image_path):
    if yolo_model is None:
        return [], None

    try:
        results = yolo_model(image_path, verbose=False)
        r = results[0]

        detections = []
        if r.boxes is not None and len(r.boxes) > 0:
            for box in r.boxes:
                cls_id = int(box.cls[0].item())
                conf = float(box.conf[0].item())
                label = r.names.get(cls_id, str(cls_id))
                detections.append((label, conf))

        annotated = r.plot()
        annotated = annotated[:, :, ::-1]  # BGR -> RGB
        return detections, annotated
    except:
        return [], None

def suspicious_manual_gps(lat, lon):
    if lat is None or lon is None:
        return False

    # exact integer coordinates often indicate dummy/test
    if abs(lat - round(lat)) < 1e-9 and abs(lon - round(lon)) < 1e-9:
        return True

    # invalid bounds
    if not (-90 <= lat <= 90 and -180 <= lon <= 180):
        return True

    return False

def should_show_map(lat, lon, gps_source):
    if lat is None or lon is None:
        return False
    if gps_source == "manual_override" and suspicious_manual_gps(lat, lon):
        return False
    return True
```

```python
def build_final_hypotheses(caption, generic_ranked, landmark_ranked, detections):
    scores = defaultdict(float)
    evidence = defaultdict(list)

    # Landmark prompts get stronger weight
    for label, score in landmark_ranked[:5]:
        scores[label] += score * 6.0
        evidence[label].append(f"Landmark CLIP {score:.3f}")

    # Generic prompts support context
    for label, score in generic_ranked[:5]:
        scores[label] += score * 3.0
        evidence[label].append(f"Place CLIP {score:.3f}")

    cap = (caption or "").lower()

    def boost(target, amount, reason):
        scores[target] += amount
        evidence[target].append(reason)

    # Caption-based heuristics
    if any(k in cap for k in ["arch", "monument", "historic", "building"]):
        boost("a historic monument", 1.5, "Caption monument/building")
        boost("a famous landmark", 1.2, "Caption landmark cue")

    if any(k in cap for k in ["boat", "boats", "harbor", "waterfront", "sea", "coast", "shore"]):
        boost("a harbor waterfront", 1.5, "Caption waterfront cue")
        boost("a marina with boats", 1.2, "Caption boats cue")
        boost("a seafront monument", 1.4, "Caption sea+monument cue")
        boost("a famous Indian monument by the sea", 1.8, "Caption sea landmark cue")
        boost("Apollo Bunder waterfront in Mumbai", 1.8, "Caption Mumbai waterfront cue")

    if "hotel" in cap:
        boost("the Taj Mahal Palace Hotel in Mumbai", 2.0, "Caption hotel cue")

    # YOLO-based heuristics
    labels = [d[0] for d in detections]
    counts = Counter(labels)

    if counts.get("boat", 0) > 0:
        boost("a harbor waterfront", 2.0, f"YOLO boat x{counts['boat']}")
        boost("a marina with boats", 1.5, f"YOLO boat x{counts['boat']}")
        boost("Apollo Bunder waterfront in Mumbai", 1.8, f"YOLO boat x{counts['boat']}")

    if counts.get("person", 0) > 0:
        boost("a famous tourist landmark", 0.8, f"YOLO person x{counts['person']}")

    ranked = sorted(scores.items(), key=lambda x: x[1], reverse=True)

    top3 = []
    if ranked:
        best_score = max(ranked[0][1], 1e-6)
        for i, (label, score) in enumerate(ranked[:3], 1):
            conf = min(99.0, round((score / best_score) * 100, 2))
            top3.append({
                "rank": i,
                "label": label,
                "confidence": conf,
                "evidence": evidence[label][:5]
            })

    return top3


    
```

```python
def create_map(lat, lon, popup_text="Detected Location"):
    m = folium.Map(location=[lat, lon], zoom_start=15)
    folium.Marker(
        [lat, lon],
        popup=popup_text,
        tooltip=popup_text
    ).add_to(m)
    return m

def ensure_csv_header(csv_path):
    if not os.path.exists(csv_path):
        with open(csv_path, "w", newline="", encoding="utf-8") as f:
            writer = csv.writer(f)
            writer.writerow([
                "timestamp",
                "image_name",
                "caption",
                "gps_source",
                "lat",
                "lon",
                "reverse_geocode",
                "top1_label",
                "top1_confidence",
                "top2_label",
                "top2_confidence",
                "top3_label",
                "top3_confidence"
            ])

def append_log(csv_path, image_name, caption, gps_source, lat, lon, reverse_geo, top3):
    ensure_csv_header(csv_path)

    row = [
        datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        image_name,
        caption if caption else "",
        gps_source,
        lat if lat is not None else "",
        lon if lon is not None else "",
        reverse_geo if reverse_geo else ""
    ]

    for i in range(3):
        if i < len(top3):
            row.extend([top3[i]["label"], top3[i]["confidence"]])
        else:
            row.extend(["", ""])

    with open(csv_path, "a", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        writer.writerow(row)

def save_report_txt(report_text, image_name):
    safe_name = os.path.splitext(os.path.basename(image_name))[0]
    report_path = os.path.join(REPORTS_DIR, f"{safe_name}_report.txt")
    with open(report_path, "w", encoding="utf-8") as f:
        f.write(report_text)
    return report_path

def build_report_text(image_name, caption, gps_source, lat, lon, reverse_geo, detections, top3):
    lines = []
    lines.append("Geo-Inference Report")
    lines.append("=" * 60)
    lines.append(f"Timestamp: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
    lines.append(f"Image: {image_name}")
    lines.append("")

    lines.append("1. Visual Understanding")
    lines.append(f"- BLIP Caption: {caption if caption else 'Unavailable'}")

    if detections:
        det_counts = Counter([d[0] for d in detections])
        det_text = ", ".join([f"{k} x{v}" for k, v in det_counts.items()])
        lines.append(f"- YOLO Objects: {det_text}")
    else:
        lines.append("- YOLO Objects: None / unavailable")

    lines.append("")
    lines.append("2. Geospatial Inputs")
    lines.append(f"- GPS Source: {gps_source}")
    lines.append(f"- Coordinates: {lat if lat is not None else 'N/A'}, {lon if lon is not None else 'N/A'}")
    lines.append(f"- Reverse Geocode: {reverse_geo if reverse_geo else 'N/A'}")

    lines.append("")
    lines.append("3. Top-3 Place Hypotheses")
    if top3:
        for h in top3:
            lines.append(f"- #{h['rank']} {h['label']} ({h['confidence']}%)")
            for ev in h["evidence"]:
                lines.append(f"    • {ev}")
    else:
        lines.append("- No robust hypothesis generated")

    lines.append("")
    lines.append("4. Final Summary")
    if top3:
        lines.append(f"Most likely location hypothesis: {top3[0]['label']} ({top3[0]['confidence']}%)")
    else:
        lines.append("Insufficient evidence to generate a robust hypothesis.")

    return "\n".join(lines)
```

```python
def analyze_photo_pro(image_path, manual_lat=None, manual_lon=None, save_outputs=True):
    result = {}
    image_name = os.path.basename(image_path)

    # -----------------------------
    # Load image
    # -----------------------------
    image = Image.open(image_path).convert("RGB")
    result["image"] = image

    display(HTML(f"<h2>🖼️ Analyzing: {image_name}</h2>"))

    plt.figure(figsize=(10, 6))
    plt.imshow(image)
    plt.axis("off")
    plt.title("Original Image")
    plt.show()

    # -----------------------------
    # 1) Caption
    # -----------------------------
    display(HTML("<h3>⏳ Step 1: BLIP Captioning...</h3>"))
    caption = generate_caption(image)
    result["caption"] = caption

    # -----------------------------
    # 2) Generic place scoring
    # -----------------------------
    display(HTML("<h3>⏳ Step 2: Generic Place Matching...</h3>"))
    generic_ranked = clip_rank_prompts(image, GENERIC_PLACE_PROMPTS)
    result["generic_ranked"] = generic_ranked

    # -----------------------------
    # 3) Landmark scoring
    # -----------------------------
    display(HTML("<h3>⏳ Step 3: Landmark Matching...</h3>"))
    landmark_ranked = clip_rank_prompts(image, LANDMARK_PROMPTS)
    result["landmark_ranked"] = landmark_ranked

    # -----------------------------
    # 4) YOLO detection
    # -----------------------------
    display(HTML("<h3>⏳ Step 4: YOLO Detection...</h3>"))
    detections, annotated = run_yolo(image_path)
    result["detections"] = detections
    result["annotated"] = annotated

    if annotated is not None:
        plt.figure(figsize=(10, 6))
        plt.imshow(annotated)
        plt.axis("off")
        plt.title("YOLO Detections")
        plt.show()

    # -----------------------------
    # 5) GPS handling
    # -----------------------------
    exif_lat, exif_lon = extract_exif_gps(image_path)

    gps_source = "none"
    final_lat, final_lon = None, None

    if exif_lat is not None and exif_lon is not None:
        final_lat, final_lon = exif_lat, exif_lon
        gps_source = "exif"
    elif manual_lat is not None and manual_lon is not None:
        final_lat, final_lon = manual_lat, manual_lon
        gps_source = "manual_override"

    result["gps_source"] = gps_source
    result["lat"] = final_lat
    result["lon"] = final_lon

    # Reverse geocode (only as support)
    reverse_geo = None
    if final_lat is not None and final_lon is not None:
        display(HTML("<h3>⏳ Step 5: Reverse Geocoding...</h3>"))
        reverse_geo = reverse_geocode(final_lat, final_lon)

    result["reverse_geo"] = reverse_geo

    # -----------------------------
    # 6) Final fusion
    # -----------------------------
    display(HTML("<h3>⏳ Step 6: Building Final Hypotheses...</h3>"))
    top3 = build_final_hypotheses(
        caption=caption,
        generic_ranked=generic_ranked,
        landmark_ranked=landmark_ranked,
        detections=detections
    )
    result["top3"] = top3

    # -----------------------------
    # 7) Display summary
    # -----------------------------
    display(HTML("<h2>📌 Final Result</h2>"))
    display(HTML(f"<p><b>📝 Caption:</b> {caption if caption else 'Unavailable'}</p>"))

    # GPS section
    if gps_source != "none":
        display(HTML(f"<p><b>📍 GPS Source:</b> {gps_source}</p>"))
        display(HTML(f"<p><b>📍 Coordinates:</b> {final_lat}, {final_lon}</p>"))

        if reverse_geo:
            display(HTML(f"<p><b>🌍 Reverse Geocode:</b> {reverse_geo}</p>"))

        if gps_source == "manual_override" and suspicious_manual_gps(final_lat, final_lon):
            display(HTML("""
            <div style='padding:10px;border:1px solid orange;border-radius:10px;background:#fff8e1;color:#8a5300;'>
                ⚠️ <b>Warning:</b> Manual GPS looks like dummy / rough coordinates.  
                Map will be skipped. Visual AI result should be preferred.
            </div>
            """))
    else:
        display(HTML("<p><b>📍 GPS:</b> Not available</p>"))

    # Generic top-3
    display(HTML("<h3>🌍 Generic Place Guesses</h3>"))
    for i, (place, score) in enumerate(generic_ranked[:3], 1):
        display(HTML(f"<p><b>{i}.</b> {place} — <b>{score*100:.2f}%</b></p>"))

    # Landmark top-5
    display(HTML("<h3>🏛️ Landmark Guesses</h3>"))
    for i, (place, score) in enumerate(landmark_ranked[:5], 1):
        display(HTML(f"<p><b>{i}.</b> {place} — <b>{score*100:.2f}%</b></p>"))

    # YOLO summary
    if detections:
        det_counts = Counter([d[0] for d in detections])
        det_text = ", ".join([f"{k} x{v}" for k, v in det_counts.items()])
        display(HTML(f"<p><b>🎯 YOLO Objects:</b> {det_text}</p>"))
    else:
        display(HTML("<p><b>🎯 YOLO Objects:</b> None / unavailable</p>"))

    # Top-3 cards
    display(HTML("<h3>🧭 Top-3 Final Location Hypotheses</h3>"))

    if top3:
        colors = ["#dff6dd", "#fff4cc", "#ffe0e0"]
        cards = "<div style='display:flex;gap:12px;flex-wrap:wrap;'>"
        for i, h in enumerate(top3):
            ev = "".join([f"<li>{e}</li>" for e in h["evidence"]])
            cards += f"""
            <div style='flex:1;min-width:260px;background:{colors[i]};border:1px solid #ccc;border-radius:16px;padding:14px;'>
                <h3>#{h['rank']} {h['label']}</h3>
                <div style='font-size:22px;font-weight:bold;'>{h['confidence']}%</div>
                <ul>{ev}</ul>
            </div>
            """
        cards += "</div>"
        display(HTML(cards))
    else:
        display(HTML("<p>No robust hypotheses generated.</p>"))

    # Final human-friendly answer
    if top3:
        best = top3[0]["label"]
        conf = top3[0]["confidence"]

        display(HTML(f"""
        <div style="padding:14px; border:1px solid #999; border-radius:12px; background:#f9f9f9; margin-top:15px;">
            <h3>✅ Final Human-Friendly Answer</h3>
            <p><b>Caption:</b> {caption if caption else 'Photo shows a landmark / place scene.'}</p>
            <p><b>Likely Location:</b> {best}</p>
            <p><b>Confidence:</b> {conf}%</p>
            <p><b>Summary:</b> Ye photo <b>{best}</b> jaisi jagah ka lag raha hai.</p>
        </div>
        """))

    # -----------------------------
    # 8) Map (only if GPS valid enough)
    # -----------------------------
    if should_show_map(final_lat, final_lon, gps_source):
        display(HTML("<h3>🗺️ Map</h3>"))
        popup_text = reverse_geo if reverse_geo else "Detected Location"
        fmap = create_map(final_lat, final_lon, popup_text=popup_text)
        display(fmap)
        result["map_shown"] = True
    else:
        if gps_source != "none":
            display(HTML("<p><b>🗺️ Map:</b> Skipped because GPS is missing / suspicious.</p>"))
        result["map_shown"] = False

    # -----------------------------
    # 9) Save CSV + report
    # -----------------------------
    report_text = build_report_text(
        image_name=image_name,
        caption=caption,
        gps_source=gps_source,
        lat=final_lat,
        lon=final_lon,
        reverse_geo=reverse_geo,
        detections=detections,
        top3=top3
    )

    result["report_text"] = report_text

    if save_outputs:
        append_log(
            csv_path=LOG_CSV_PATH,
            image_name=image_name,
            caption=caption,
            gps_source=gps_source,
            lat=final_lat,
            lon=final_lon,
            reverse_geo=reverse_geo,
            top3=top3
        )

        report_path = save_report_txt(report_text, image_name)
        result["report_path"] = report_path

        display(HTML("<h3>📄 Auto-generated Report</h3>"))
        print(report_text)
        display(HTML(f"<p><b>CSV Log saved:</b> {LOG_CSV_PATH}</p>"))
        display(HTML(f"<p><b>Text Report saved:</b> {report_path}</p>"))

    return result
```

```python
import os
import io
import csv
from datetime import datetime
from collections import Counter, defaultdict
from urllib.parse import urlencode

import numpy as np
import pandas as pd
import requests
from PIL import Image, ExifTags

import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

import torch
import open_clip

# Optional YOLO
try:
    from ultralytics import YOLO
    HAS_YOLO = True
except:
    HAS_YOLO = False

# =========================
# SET YOUR GOOGLE MAPS API KEY HERE
# =========================
GOOGLE_MAPS_API_KEY = "PASTE_YOUR_GOOGLE_MAPS_API_KEY_HERE"

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", DEVICE)

# Load CLIP
print("⏳ Loading CLIP...")
clip_model, _, clip_preprocess = open_clip.create_model_and_transforms(
    "ViT-B-32",
    pretrained="laion2b_s34b_b79k",
    device=DEVICE
)
clip_tokenizer = open_clip.get_tokenizer("ViT-B-32")
print("✅ CLIP loaded")

# Load YOLO
yolo_model = None
if HAS_YOLO:
    try:
        print("⏳ Loading YOLO...")
        yolo_model = YOLO("yolov8n.pt")
        print("✅ YOLO loaded")
    except Exception as e:
        print("⚠️ YOLO failed:", e)
        yolo_model = None
else:
    print("⚠️ YOLO not available")

LOG_CSV_PATH = "googlemaps_geo_log.csv"
REPORTS_DIR = "googlemaps_reports"
os.makedirs(REPORTS_DIR, exist_ok=True)
```

```python
GENERIC_PLACE_PROMPTS = [
    "a busy city street",
    "a downtown urban area",
    "a residential neighborhood",
    "a suburban street",
    "a shopping mall exterior",
    "a marketplace in a city",
    "a railway station area",
    "an airport terminal exterior",
    "a mountain landscape",
    "a forest trail",
    "a desert scene",
    "a beach coastline",
    "a lakeside scene",
    "a riverfront area",
    "a tourist attraction",
    "a historic monument",
    "a temple complex",
    "a church exterior",
    "a mosque exterior",
    "a castle exterior",
    "an ancient ruin",
    "a famous landmark",
    "a highway roadside",
    "a bridge crossing",
    "a hotel exterior",
    "a resort entrance",
    "a restaurant streetfront",
    "a cafe outdoor seating area",
    "a park with walking paths",
    "a sports stadium exterior",
    "a coastal town",
    "a village center",
    "a harbor waterfront",
    "a marina with boats",
    "a seafront monument"
]

LANDMARK_PROMPTS = [
    "the Gateway of India monument in Mumbai",
    "the Taj Mahal Palace Hotel in Mumbai",
    "Apollo Bunder waterfront in Mumbai",
    "a historic waterfront monument in Mumbai",
    "a famous tourist landmark in Mumbai",
    "a harbor front arch monument in India",
    "a coastal heritage monument in India",
    "a waterfront monument with boats in Mumbai",
    "a colonial era monument in Mumbai",
    "a famous Indian monument by the sea",
    "a tourist place in South Mumbai",
    "a seafront landmark in Mumbai",
    "the India Gate monument in New Delhi",
    "the Taj Mahal in Agra",
    "the Qutub Minar in Delhi",
    "the Red Fort in Delhi",
    "the Charminar in Hyderabad",
    "the Victoria Memorial in Kolkata",
    "the Howrah Bridge in Kolkata",
    "the Mysore Palace in Mysuru",
    "the Hawa Mahal in Jaipur",
    "the Amer Fort in Jaipur",
    "the Golden Temple in Amritsar",
    "the Marine Drive seafront in Mumbai"
]
```

```python
def dms_to_decimal(dms, ref):
    try:
        degrees = float(dms[0][0]) / float(dms[0][1])
        minutes = float(dms[1][0]) / float(dms[1][1])
        seconds = float(dms[2][0]) / float(dms[2][1])
        dec = degrees + minutes / 60 + seconds / 3600
        if ref in ['S', 'W']:
            dec = -dec
        return dec
    except:
        return None

def extract_exif_gps(image_path):
    try:
        img = Image.open(image_path)
        exif_data = img._getexif()
        if not exif_data:
            return None, None

        exif = {}
        for tag_id, value in exif_data.items():
            tag = ExifTags.TAGS.get(tag_id, tag_id)
            exif[tag] = value

        gps_info = exif.get("GPSInfo", None)
        if not gps_info:
            return None, None

        gps_parsed = {}
        for key in gps_info.keys():
            decoded = ExifTags.GPSTAGS.get(key, key)
            gps_parsed[decoded] = gps_info[key]

        if all(k in gps_parsed for k in ["GPSLatitude", "GPSLatitudeRef", "GPSLongitude", "GPSLongitudeRef"]):
            lat = dms_to_decimal(gps_parsed["GPSLatitude"], gps_parsed["GPSLatitudeRef"])
            lon = dms_to_decimal(gps_parsed["GPSLongitude"], gps_parsed["GPSLongitudeRef"])
            return lat, lon

        return None, None
    except:
        return None, None

def suspicious_manual_gps(lat, lon):
    if lat is None or lon is None:
        return False
    if abs(lat - round(lat)) < 1e-9 and abs(lon - round(lon)) < 1e-9:
        return True
    if not (-90 <= lat <= 90 and -180 <= lon <= 180):
        return True
    return False

def google_reverse_geocode(lat, lon, api_key):
    if not api_key or "PASTE_YOUR" in api_key:
        return None, None

    url = "https://maps.googleapis.com/maps/api/geocode/json"
    params = {
        "latlng": f"{lat},{lon}",
        "key": api_key
    }

    try:
        r = requests.get(url, params=params, timeout=10)
        r.raise_for_status()
        data = r.json()

        if data.get("status") == "OK" and data.get("results"):
            formatted = data["results"][0]["formatted_address"]
            place_id = data["results"][0].get("place_id")
            return formatted, place_id

        return None, None
    except:
        return None, None

def google_nearby_places(lat, lon, api_key, radius=500, place_type="tourist_attraction"):
    if not api_key or "PASTE_YOUR" in api_key:
        return []

    url = "https://maps.googleapis.com/maps/api/place/nearbysearch/json"
    params = {
        "location": f"{lat},{lon}",
        "radius": radius,
        "type": place_type,
        "key": api_key
    }

    try:
        r = requests.get(url, params=params, timeout=10)
        r.raise_for_status()
        data = r.json()

        if data.get("status") in ["OK", "ZERO_RESULTS"]:
            return data.get("results", [])
        return []
    except:
        return []

def google_static_map_url(lat, lon, api_key, zoom=16, size="640x400", maptype="roadmap"):
    if not api_key or "PASTE_YOUR" in api_key:
        return None

    base = "https://maps.googleapis.com/maps/api/staticmap"
    params = {
        "center": f"{lat},{lon}",
        "zoom": zoom,
        "size": size,
        "maptype": maptype,
        "markers": f"color:red|label:P|{lat},{lon}",
        "key": api_key
    }
    return f"{base}?{urlencode(params)}"
```

```python
def clip_rank_prompts(image, prompts):
    try:
        image_tensor = clip_preprocess(image).unsqueeze(0).to(DEVICE)
        text_tokens = clip_tokenizer(prompts).to(DEVICE)

        with torch.no_grad():
            image_features = clip_model.encode_image(image_tensor)
            text_features = clip_model.encode_text(text_tokens)

            image_features /= image_features.norm(dim=-1, keepdim=True)
            text_features /= text_features.norm(dim=-1, keepdim=True)

            probs = (100.0 * image_features @ text_features.T).softmax(dim=-1)[0].cpu().numpy()

        ranked = sorted(zip(prompts, probs), key=lambda x: x[1], reverse=True)
        return ranked
    except:
        return []

def run_yolo(image_path):
    if yolo_model is None:
        return [], None

    try:
        results = yolo_model(image_path, verbose=False)
        r = results[0]

        detections = []
        if r.boxes is not None and len(r.boxes) > 0:
            for box in r.boxes:
                cls_id = int(box.cls[0].item())
                conf = float(box.conf[0].item())
                label = r.names.get(cls_id, str(cls_id))
                detections.append((label, conf))

        annotated = r.plot()
        annotated = annotated[:, :, ::-1]
        return detections, annotated
    except:
        return [], None

def make_pseudo_caption(top_generic, top_landmark, detections):
    parts = []
    if top_landmark:
        parts.append(f"Likely landmark: {top_landmark[0][0]}")
    if top_generic:
        parts.append(f"Scene: {top_generic[0][0]}")
    if detections:
        det_counts = Counter([d[0] for d in detections])
        det_text = ", ".join([f"{k} x{v}" for k, v in det_counts.most_common(4)])
        parts.append(f"Objects: {det_text}")
    if not parts:
        return "Landmark/place scene detected"
    return " | ".join(parts)

def build_final_hypotheses(generic_ranked, landmark_ranked, detections):
    scores = defaultdict(float)
    evidence = defaultdict(list)

    for label, score in landmark_ranked[:5]:
        scores[label] += score * 6.0
        evidence[label].append(f"Landmark CLIP {score:.3f}")

    for label, score in generic_ranked[:5]:
        scores[label] += score * 3.0
        evidence[label].append(f"Place CLIP {score:.3f}")

    labels = [d[0] for d in detections]
    counts = Counter(labels)

    def boost(target, amount, reason):
        scores[target] += amount
        evidence[target].append(reason)

    if counts.get("boat", 0) > 0:
        boost("Apollo Bunder waterfront in Mumbai", 2.0, f"YOLO boat x{counts['boat']}")
        boost("a harbor waterfront", 1.5, f"YOLO boat x{counts['boat']}")
        boost("a marina with boats", 1.2, f"YOLO boat x{counts['boat']}")

    if counts.get("person", 0) > 0:
        boost("a famous tourist landmark", 0.8, f"YOLO person x{counts['person']}")

    ranked = sorted(scores.items(), key=lambda x: x[1], reverse=True)

    top3 = []
    if ranked:
        best_score = max(ranked[0][1], 1e-6)
        for i, (label, score) in enumerate(ranked[:3], 1):
            conf = min(99.0, round((score / best_score) * 100, 2))
            top3.append({
                "rank": i,
                "label": label,
                "confidence": conf,
                "evidence": evidence[label][:5]
            })

    return top3
```

```python
def ensure_csv_header(csv_path):
    if not os.path.exists(csv_path):
        with open(csv_path, "w", newline="", encoding="utf-8") as f:
            writer = csv.writer(f)
            writer.writerow([
                "timestamp", "image_name", "caption",
                "gps_source", "lat", "lon", "reverse_geocode",
                "top1_label", "top1_confidence",
                "top2_label", "top2_confidence",
                "top3_label", "top3_confidence"
            ])

def append_log(csv_path, image_name, caption, gps_source, lat, lon, reverse_geo, top3):
    ensure_csv_header(csv_path)
    row = [
        datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        image_name,
        caption,
        gps_source,
        lat if lat is not None else "",
        lon if lon is not None else "",
        reverse_geo if reverse_geo else ""
    ]
    for i in range(3):
        if i < len(top3):
            row.extend([top3[i]["label"], top3[i]["confidence"]])
        else:
            row.extend(["", ""])
    with open(csv_path, "a", newline="", encoding="utf-8") as f:
        csv.writer(f).writerow(row)

def save_report_txt(report_text, image_name):
    safe_name = os.path.splitext(os.path.basename(image_name))[0]
    path = os.path.join(REPORTS_DIR, f"{safe_name}_report.txt")
    with open(path, "w", encoding="utf-8") as f:
        f.write(report_text)
    return path
```

```python
def analyze_photo_googlemaps(image_path, manual_lat=None, manual_lon=None):
    image_name = os.path.basename(image_path)
    image = Image.open(image_path).convert("RGB")

    display(HTML(f"<h2>🖼️ Analyzing (Google Maps): {image_name}</h2>"))

    plt.figure(figsize=(10, 6))
    plt.imshow(image)
    plt.axis("off")
    plt.title("Original Image")
    plt.show()

    # 1) CLIP generic
    display(HTML("<h3>⏳ Step 1: Generic Place Matching...</h3>"))
    generic_ranked = clip_rank_prompts(image, GENERIC_PLACE_PROMPTS)

    # 2) CLIP landmark
    display(HTML("<h3>⏳ Step 2: Landmark Matching...</h3>"))
    landmark_ranked = clip_rank_prompts(image, LANDMARK_PROMPTS)

    # 3) YOLO
    display(HTML("<h3>⏳ Step 3: YOLO Detection...</h3>"))
    detections, annotated = run_yolo(image_path)

    if annotated is not None:
        plt.figure(figsize=(10, 6))
        plt.imshow(annotated)
        plt.axis("off")
        plt.title("YOLO Detections")
        plt.show()

    # 4) GPS
    exif_lat, exif_lon = extract_exif_gps(image_path)
    gps_source = "none"
    final_lat, final_lon = None, None

    if exif_lat is not None and exif_lon is not None:
        final_lat, final_lon = exif_lat, exif_lon
        gps_source = "exif"
    elif manual_lat is not None and manual_lon is not None:
        final_lat, final_lon = manual_lat, manual_lon
        gps_source = "manual_override"

    # 5) Google reverse geocode + nearby places
    reverse_geo = None
    place_id = None
    nearby_places = []

    if final_lat is not None and final_lon is not None:
        display(HTML("<h3>⏳ Step 4: Google Reverse Geocoding...</h3>"))
        reverse_geo, place_id = google_reverse_geocode(final_lat, final_lon, GOOGLE_MAPS_API_KEY)

        display(HTML("<h3>⏳ Step 5: Google Nearby Places...</h3>"))
        nearby_places = google_nearby_places(final_lat, final_lon, GOOGLE_MAPS_API_KEY, radius=500, place_type="tourist_attraction")

    # 6) Pseudo caption
    caption = make_pseudo_caption(generic_ranked, landmark_ranked, detections)

    # 7) Final fusion
    display(HTML("<h3>⏳ Step 6: Final Fusion...</h3>"))
    top3 = build_final_hypotheses(generic_ranked, landmark_ranked, detections)

    # 8) Output
    display(HTML("<h2>📌 Final Result</h2>"))
    display(HTML(f"<p><b>📝 Caption:</b> {caption}</p>"))

    if gps_source != "none":
        display(HTML(f"<p><b>📍 GPS Source:</b> {gps_source}</p>"))
        display(HTML(f"<p><b>📍 Coordinates:</b> {final_lat}, {final_lon}</p>"))

        if reverse_geo:
            display(HTML(f"<p><b>🌍 Google Reverse Geocode:</b> {reverse_geo}</p>"))

        if gps_source == "manual_override" and suspicious_manual_gps(final_lat, final_lon):
            display(HTML("""
            <div style='padding:10px;border:1px solid orange;border-radius:10px;background:#fff8e1;color:#8a5300;'>
                ⚠️ Warning: Manual GPS looks suspicious. Visual result should be preferred.
            </div>
            """))
    else:
        display(HTML("<p><b>📍 GPS:</b> Not available</p>"))

    # Nearby places
    if nearby_places:
        display(HTML("<h3>📍 Google Nearby Places (Tourist Attractions)</h3>"))
        for i, p in enumerate(nearby_places[:5], 1):
            name = p.get("name", "Unknown")
            vicinity = p.get("vicinity", "")
            rating = p.get("rating", "N/A")
            display(HTML(f"<p><b>{i}.</b> {name} — {vicinity} (Rating: {rating})</p>"))

    # Generic
    display(HTML("<h3>🌍 Generic Place Guesses</h3>"))
    for i, (place, score) in enumerate(generic_ranked[:3], 1):
        display(HTML(f"<p><b>{i}.</b> {place} — <b>{score*100:.2f}%</b></p>"))

    # Landmark
    display(HTML("<h3>🏛️ Landmark Guesses</h3>"))
    for i, (place, score) in enumerate(landmark_ranked[:5], 1):
        display(HTML(f"<p><b>{i}.</b> {place} — <b>{score*100:.2f}%</b></p>"))

    # YOLO
    if detections:
        det_counts = Counter([d[0] for d in detections])
        det_text = ", ".join([f"{k} x{v}" for k, v in det_counts.items()])
        display(HTML(f"<p><b>🎯 YOLO Objects:</b> {det_text}</p>"))
    else:
        display(HTML("<p><b>🎯 YOLO Objects:</b> None / unavailable</p>"))

    # Top-3
    display(HTML("<h3>🧭 Top-3 Final Location Hypotheses</h3>"))
    if top3:
        colors = ["#dff6dd", "#fff4cc", "#ffe0e0"]
        cards = "<div style='display:flex;gap:12px;flex-wrap:wrap;'>"
        for i, h in enumerate(top3):
            ev = "".join([f"<li>{e}</li>" for e in h["evidence"]])
            cards += f"""
            <div style='flex:1;min-width:260px;background:{colors[i]};border:1px solid #ccc;border-radius:16px;padding:14px;'>
                <h3>#{h['rank']} {h['label']}</h3>
                <div style='font-size:22px;font-weight:bold;'>{h['confidence']}%</div>
                <ul>{ev}</ul>
            </div>
            """
        cards += "</div>"
        display(HTML(cards))

        best = top3[0]["label"]
        conf = top3[0]["confidence"]

        display(HTML(f"""
        <div style="padding:14px; border:1px solid #999; border-radius:12px; background:#f9f9f9; margin-top:15px;">
            <h3>✅ Final Human-Friendly Answer</h3>
            <p><b>Caption:</b> {caption}</p>
            <p><b>Likely Location:</b> {best}</p>
            <p><b>Confidence:</b> {conf}%</p>
            <p><b>Summary:</b> Ye photo <b>{best}</b> jaisi jagah ka lag raha hai.</p>
        </div>
        """))

    # 9) Google Static Map
    if final_lat is not None and final_lon is not None and not (gps_source == "manual_override" and suspicious_manual_gps(final_lat, final_lon)):
        map_url = google_static_map_url(final_lat, final_lon, GOOGLE_MAPS_API_KEY)
        if map_url:
            display(HTML("<h3>🗺️ Google Static Map</h3>"))
            display(HTML(f'<img src="{map_url}" style="max-width:100%; border:1px solid #ccc; border-radius:8px;">'))
    elif gps_source != "none":
        display(HTML("<p><b>🗺️ Map:</b> Skipped because GPS is suspicious / missing.</p>"))

    # 10) Save outputs
    append_log(LOG_CSV_PATH, image_name, caption, gps_source, final_lat, final_lon, reverse_geo, top3)

    report_lines = [
        f"Image: {image_name}",
        f"Caption: {caption}",
        f"GPS Source: {gps_source}",
        f"Coordinates: {final_lat}, {final_lon}",
        f"Google Reverse Geocode: {reverse_geo}",
        "Top-3 Hypotheses:"
    ]
    for h in top3:
        report_lines.append(f"  #{h['rank']} {h['label']} ({h['confidence']}%)")

    if nearby_places:
        report_lines.append("Nearby Places:")
        for p in nearby_places[:5]:
            report_lines.append(f"  - {p.get('name', 'Unknown')} | {p.get('vicinity', '')}")

    report_text = "\n".join(report_lines)
    report_path = save_report_txt(report_text, image_name)

    display(HTML(f"<p><b>CSV Log saved:</b> {LOG_CSV_PATH}</p>"))
    display(HTML(f"<p><b>Text Report saved:</b> {report_path}</p>"))
```

```python
upload = widgets.FileUpload(
    accept='image/*',
    multiple=False,
    description='Upload Photo'
)

manual_lat_input = widgets.Text(
    value="",
    placeholder="Optional (e.g. 18.9220)",
    description="Lat:"
)

manual_lon_input = widgets.Text(
    value="",
    placeholder="Optional (e.g. 72.8347)",
    description="Lon:"
)

analyze_btn = widgets.Button(
    description='Analyze (Google Maps)',
    button_style='success',
    icon='search'
)

output = widgets.Output()

def parse_float_safe(x):
    x = x.strip()
    if x == "":
        return None
    try:
        return float(x)
    except:
        return None

def on_analyze_click(b):
    with output:
        clear_output()

        if not upload.value:
            display(HTML("<b style='color:red;'>❌ Pehle photo upload karo.</b>"))
            return

        if isinstance(upload.value, tuple):
            uploaded_file = upload.value[0]
            file_name = uploaded_file['name']
            content = uploaded_file['content']
        else:
            uploaded_file = list(upload.value.values())[0]
            file_name = uploaded_file['metadata']['name']
            content = uploaded_file['content']

        temp_path = file_name
        with open(temp_path, "wb") as f:
            f.write(content)

        manual_lat = parse_float_safe(manual_lat_input.value)
        manual_lon = parse_float_safe(manual_lon_input.value)

        analyze_photo_googlemaps(temp_path, manual_lat, manual_lon)

analyze_btn.on_click(on_analyze_click)

display(HTML("<h2>🌍 Google Maps Version: Landmark Guess + Reverse Geocode + Nearby Places + Static Map</h2>"))
display(HTML("<p>Upload photo karo. GPS optional hai. Agar exact location nahi pata toh blank chhodo.</p>"))
display(upload)
display(widgets.HBox([manual_lat_input, manual_lon_input]))
display(analyze_btn)
display(output)
```
